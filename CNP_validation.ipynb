# Isotropic non-Poisson source-clustering validation

This standalone notebook preserves the optional $C_{\rm NP}\equiv C_{{\rm NP},0}$ validation used in Appendix A. It intentionally contains no spiral-harmonic $C_{{\rm NP},N}$ term. Each realization is residualized by subtracting the deterministic local-WKBJ axisymmetric ensemble mean used by the analytic prediction, rather than a radial profile re-estimated from that realization. The two-point estimators enumerate every unordered pair of pixels in the analysis window; no random pair subsampling or field interpolation is used. Analytic predictions are averaged over the same finite separation bins with the planar $r\,dr$ pair measure. Here an event kernel means the Gaussian spatial profile of one injection event. Narrow event kernels are integrated over square pixel faces, while broad event kernels are evaluated by convolution on the pixel-centre mesh. This division is purely numerical, not physical; both the simulated field and every analytic covariance component, including $C_{\rm NP}$, target the same square-pixel observable. Only complete-azimuth statistics are interpreted as tests of $C_{{\rm NP},0}$; the arm-phase plot is retained as a diagnostic outside that validation claim.


```python
# %% Block 1: Import packages and set global parameters
import numpy as np
from numpy.random import default_rng
from scipy.special import exp1, ndtr
from scipy.integrate import quad
from scipy.ndimage import gaussian_filter
from scipy.stats import qmc
import matplotlib.pyplot as plt
import os

import warnings
warnings.filterwarnings('ignore')

# Controlled non-uniform validation setup; these geometry and source-profile
# choices are designed to exercise the analytic terms, not to fit a galaxy.
R_in, R_out = 3.0, 15

# Diffusion parameters.  These control the KTZ smoothing scale,
# ell_ast = sqrt(sigma_scalar^2 + 2 kappa t_star), rather than a direct observable.
kappa = 0.3579         # kpc^2/Gyr (same as KT25)
sigma_scalar = 0.06  # initial Gaussian injection dispersion (kpc)
t_star = 0.50        # maximum source age (Gyr)

# Spiral structure: a single logarithmic spiral with m_arms arms and pitch angle i.
m_arms = 3
pitch_angle = 20.0   # degree
pitch_rad = np.deg2rad(pitch_angle)
Theta0 = 0.0

# Spiral arm modulation h(Theta) = sum_n g_n cos(n Theta + alpha_n).
# Keep all harmonic parameters here; the numerical source rate and KTZ theory
# blocks read these arrays directly.
eta = 0.40
harmonic_n = np.array([1, 2, 3], dtype=int)
harmonic_g = np.array([1.00, 0.40, 0.20], dtype=float)
harmonic_alpha = np.array([0.00, 0.00, 0.00], dtype=float)

def h_func(Theta):
    Theta = np.asarray(Theta)
    h_val = np.zeros_like(Theta, dtype=float)
    for n, g, alpha in zip(harmonic_n, harmonic_g, harmonic_alpha):
        h_val = h_val + g*np.cos(n*Theta + alpha)
    return h_val

Theta_check = np.linspace(0, 2*np.pi, 4096, endpoint=False)
rate_floor = 1.0 + eta*np.min(h_func(Theta_check))
if rate_floor <= 0:
    raise ValueError("Spiral harmonic settings make lambda_rate negative; lower eta or harmonic_g.")

# Axisymmetric background rate lambda_0(varpi) = lambda0_0 exp(-varpi/h_R).
# Each synthetic event carries unit yield (Y_val below), so lambda0_0 is an
# effective event rate.
lambda0_0 = 5.0e2   # kpc^-2 Gyr^-1
h_R = 4.8
validation_rmax = 5.0
# A narrow arm-phase selection is used below to keep the linear
# connected-Poisson spiral term from cancelling azimuthally.
sector_phase_halfwidth = np.deg2rad(30.0)
sector_validation_rmax = 1.5
sector_analysis_R_in = R_in + 1.5
sector_analysis_R_out = R_out - 1.5

# This standalone Appendix-A notebook always runs the isotropic C_NP0 case.
enable_source_clustering = True
validation_output_prefix = 'appendix_cnp_'

def validation_figure_path(stem):
    return f'figs/{validation_output_prefix}validation_{stem}.png'

cluster_fraction = 0.50              # fraction of events drawn from clustered complexes
cluster_radius = 0.35                # kpc, one-sigma child offset around each parent
mean_children_per_cluster = 20.0      # mean child events per parent complex
cluster_theory_nodes = 24            # quadrature nodes for the C_NP age integral

# Output (fixed)
Y_val = 1.0

# Computational grid. Simulated maps and analytic covariances both use
# the exact response of square pixels with this side length. As in the
# elemental calculation, unresolved kernels retain their individual ages
# and are integrated over pixel faces; only kernels broader than three
# pixels use logarithmic width bins and convolution on the pixel-centre mesh.
dx = 0.20
scalar_age_bins = 40
scalar_exact_sigma_threshold = 3.0 * dx
scalar_pixel_response_samples = 4096
grid_radius = R_out + 1.0

# Number of independent realizations used for the error bands.
n_realizations = 10

rng = default_rng(42)
```

```python
# %% Block 2: Auxiliary functions (spiral phase, source rate, wavenumber)
def spiral_phase(varpi, phi):
    r"""The phase of a logarithmic spiral $\Theta$."""
    Phi = (m_arms / np.tan(pitch_rad)) * np.log(varpi / R_in)
    return Phi - m_arms * phi + Theta0

def lambda_rate(varpi, phi):
    r"""Non-uniform event rate $\lambda(\varpi,\phi)$."""
    lam0 = lambda0_0 * np.exp(-varpi / h_R)
    Theta = spiral_phase(varpi, phi)
    return lam0 * (1.0 + eta * h_func(Theta))

def k_perp(varpi):
    r"""Local arm-normal wavenumber $k_\perp$."""
    dPhi = (m_arms / np.tan(pitch_rad)) / varpi
    return np.sqrt(dPhi**2 + (m_arms / varpi)**2)

# Visualising the spiral pattern and its arm-profile harmonics.
varpi_plot = np.linspace(R_in, R_out, 200)
profile_theta = np.linspace(0, 2*np.pi, 600)
profile_total = h_func(profile_theta)

fig, (ax_spiral, ax_profile) = plt.subplots(1, 2, figsize=(12, 5))
for k in range(m_arms):
    # Solve Phi(varpi) - m*phi + Theta0 = 0 -> phi = (Phi(varpi) + Theta0 + 2*pi*k)/m.
    phi_sp = ((m_arms / np.tan(pitch_rad)) * np.log(varpi_plot / R_in) + Theta0 + 2*np.pi*k) / m_arms
    ax_spiral.plot(varpi_plot * np.cos(phi_sp), varpi_plot * np.sin(phi_sp), 'k-', lw=2)
ax_spiral.set_xlim(-R_out, R_out)
ax_spiral.set_ylim(-R_out, R_out)
ax_spiral.set_aspect('equal')
ax_spiral.set_title(r'Spiral skeleton ($\Theta = 0$)')
ax_spiral.set_xlabel('x (kpc)')
ax_spiral.set_ylabel('y (kpc)')

for n, g, alpha in zip(harmonic_n, harmonic_g, harmonic_alpha):
    component = g*np.cos(n*profile_theta + alpha)
    ax_profile.plot(profile_theta, component, '--', lw=1, alpha=0.75, label=fr'$n={n}$')
ax_profile.plot(profile_theta, profile_total, 'k-', lw=2, label=r'$h(\Theta)$')
ax_profile.axhline(0, color='grey', lw=0.6)
ax_profile.set_xlim(0, 2*np.pi)
ax_profile.set_xticks([0, np.pi, 2*np.pi])
ax_profile.set_xticklabels([r'$0$', r'$\pi$', r'$2\pi$'])
ax_profile.set_xlabel(r'$\Theta$')
ax_profile.set_ylabel(r'$h(\Theta)$')
ax_profile.set_title(r'Arm profile')
ax_profile.grid(alpha=0.25)
ax_profile.legend(fontsize=8, ncol=2)

plt.tight_layout()
plt.show()
```

```python
# %% Block 3: Generating synthetic galaxies (inhomogeneous Poisson + Gaussian diffusion)
# --- Expected number of events ---
Nr, Nphi = 200, 400
varpi_edges = np.linspace(R_in, R_out, Nr+1)
phi_edges = np.linspace(0, 2*np.pi, Nphi+1)
varpi_c = 0.5*(varpi_edges[:-1] + varpi_edges[1:])
phi_c = 0.5*(phi_edges[:-1] + phi_edges[1:])
dvarpi = np.diff(varpi_edges)
dphi = 2*np.pi / Nphi
area_cells = (varpi_c * dvarpi)[:, None] * dphi
lam_grid = np.array([[lambda_rate(varpi_c[ir], phi_c[ip]) for ip in range(Nphi)] for ir in range(Nr)])
expected_total = np.sum(lam_grid * area_cells) * t_star
lam_max = np.max(lam_grid)

# --- Pixel-centre mesh for abundance fields ---
grid_x = np.arange(-grid_radius, grid_radius+dx, dx)
grid_y = np.arange(-grid_radius, grid_radius+dx, dx)
X, Y = np.meshgrid(grid_x, grid_y)


def _deposit_scalar_binned_gaussian(event_x, event_y, weight, sigma):
    """Deposit broad kernels by nearest-cell binning and convolution."""
    nx = len(grid_x)
    ny = len(grid_y)
    ix = np.floor((event_x - grid_x[0]) / dx + 0.5).astype(int)
    iy = np.floor((event_y - grid_y[0]) / dx + 0.5).astype(int)
    inside = (
        (ix >= 0) & (ix < nx) & (iy >= 0) & (iy < ny)
        & np.isfinite(weight)
    )
    source = np.bincount(
        iy[inside] * nx + ix[inside],
        weights=np.broadcast_to(weight, np.shape(event_x))[inside],
        minlength=nx * ny,
    ).reshape(ny, nx)
    return gaussian_filter(
        source, sigma=max(sigma / dx, 1.0e-3), mode='constant',
        cval=0.0, truncate=5.0,
    ) / dx**2


def _deposit_scalar_pixel_integrated(
    event_x, event_y, weight, sigma, cutoff=5.0
):
    """Integrate continuous Gaussian kernels over square pixel faces."""
    result = np.zeros_like(X, dtype=float)
    half_pixel = 0.5 * dx
    nx = len(grid_x)
    ny = len(grid_y)
    sigma = np.broadcast_to(np.asarray(sigma, dtype=float), np.shape(event_x))
    weight = np.broadcast_to(np.asarray(weight, dtype=float), np.shape(event_x))
    for x_event, y_event, event_weight, event_sigma in zip(
        event_x, event_y, weight, sigma
    ):
        if (
            not np.isfinite(event_weight)
            or not np.isfinite(event_sigma)
            or event_sigma <= 0
        ):
            continue
        support = cutoff * event_sigma + half_pixel
        ix0 = max(0, int(np.ceil((x_event - support - grid_x[0]) / dx)))
        ix1 = min(nx - 1, int(np.floor((x_event + support - grid_x[0]) / dx)))
        iy0 = max(0, int(np.ceil((y_event - support - grid_y[0]) / dx)))
        iy1 = min(ny - 1, int(np.floor((y_event + support - grid_y[0]) / dx)))
        if ix0 > ix1 or iy0 > iy1:
            continue
        pixel_x = grid_x[ix0:ix1 + 1]
        pixel_y = grid_y[iy0:iy1 + 1]
        probability_x = (
            ndtr((pixel_x + half_pixel - x_event) / event_sigma)
            - ndtr((pixel_x - half_pixel - x_event) / event_sigma)
        )
        probability_y = (
            ndtr((pixel_y + half_pixel - y_event) / event_sigma)
            - ndtr((pixel_y - half_pixel - y_event) / event_sigma)
        )
        result[iy0:iy1 + 1, ix0:ix1 + 1] += (
            event_weight * np.outer(probability_y, probability_x) / dx**2
        )
    return result


def sample_spiral_positions(n_sample, rng):
    """Sample positions from the spiral-modulated mean rate in the annulus."""
    if n_sample <= 0:
        return np.empty(0), np.empty(0)

    events_varpi = np.empty(0)
    events_phi = np.empty(0)
    n_cand = max(int(3*n_sample), 1)
    while len(events_varpi) < n_sample:
        varpi_cand = np.sqrt(rng.uniform(R_in**2, R_out**2, n_cand))
        phi_cand = rng.uniform(0, 2*np.pi, n_cand)
        lam_cand = lambda_rate(varpi_cand, phi_cand)
        accept = rng.uniform(size=n_cand) < lam_cand / lam_max
        events_varpi = np.concatenate([events_varpi, varpi_cand[accept]])
        events_phi = np.concatenate([events_phi, phi_cand[accept]])
        n_cand = max(2*(n_sample - len(events_varpi)), 1)

    return events_varpi[:n_sample], events_phi[:n_sample]

def sample_clustered_positions(rng):
    """Sample an exact Thomas component without fixing its child count."""
    mean_children = max(mean_children_per_cluster, 1.0e-6)
    expected_parent_complexes = (
        cluster_fraction * expected_total / mean_children
    )
    n_parent = rng.poisson(expected_parent_complexes)
    if n_parent == 0:
        return np.empty(0), np.empty(0)

    parent_varpi, parent_phi = sample_spiral_positions(n_parent, rng)
    parent_x = parent_varpi*np.cos(parent_phi)
    parent_y = parent_varpi*np.sin(parent_phi)
    child_counts = rng.poisson(mean_children, n_parent)
    n_child = int(np.sum(child_counts))
    if n_child == 0:
        return np.empty(0), np.empty(0)

    px = np.repeat(parent_x, child_counts)
    py = np.repeat(parent_y, child_counts)
    child_x = px + rng.normal(0.0, cluster_radius, n_child)
    child_y = py + rng.normal(0.0, cluster_radius, n_child)
    child_radius = np.hypot(child_x, child_y)
    retained = (child_radius >= R_in) & (child_radius <= R_out)
    return child_x[retained], child_y[retained]

def generate_synthetic_galaxy(rng, return_events=False):
    """Generate one enrichment realization with optional Thomas source clustering."""
    use_clustering = enable_source_clustering and cluster_fraction > 0
    if use_clustering:
        f_cl = np.clip(cluster_fraction, 0.0, 1.0)
        n_poisson = rng.poisson((1.0 - f_cl) * expected_total)
        cluster_x, cluster_y = sample_clustered_positions(rng)
    else:
        n_poisson = rng.poisson(expected_total)
        cluster_x = np.empty(0)
        cluster_y = np.empty(0)

    poisson_varpi, poisson_phi = sample_spiral_positions(n_poisson, rng)
    poisson_x = poisson_varpi*np.cos(poisson_phi)
    poisson_y = poisson_varpi*np.sin(poisson_phi)
    event_x = np.concatenate([poisson_x, cluster_x])
    event_y = np.concatenate([poisson_y, cluster_y])
    event_is_clustered = np.concatenate([
        np.zeros(len(poisson_x), dtype=bool),
        np.ones(len(cluster_x), dtype=bool),
    ])

    if len(event_x) > 0:
        order = rng.permutation(len(event_x))
        event_x = event_x[order]
        event_y = event_y[order]
        event_is_clustered = event_is_clustered[order]
    ages = rng.uniform(0, t_star, len(event_x))

    event_width2 = sigma_scalar**2 + 2 * kappa * ages
    exact_selected = (
        event_width2 < scalar_exact_sigma_threshold**2
    )
    Z = np.zeros_like(X)
    if np.any(exact_selected):
        Z += _deposit_scalar_pixel_integrated(
            event_x[exact_selected], event_y[exact_selected],
            np.full(np.sum(exact_selected), Y_val),
            np.sqrt(event_width2[exact_selected]),
        )

    width2_edges = np.geomspace(
        sigma_scalar**2, sigma_scalar**2 + 2 * kappa * t_star,
        scalar_age_bins + 1,
    )
    for bin_index, (lower, upper) in enumerate(zip(
        width2_edges[:-1], width2_edges[1:]
    )):
        selected = (
            ~exact_selected & (event_width2 >= lower)
            & (event_width2 < upper)
        )
        if not np.any(selected):
            continue
        sigma_bin = (lower * upper)**0.25
        Z += _deposit_scalar_binned_gaussian(
            event_x[selected], event_y[selected], Y_val, sigma_bin
        )

    if return_events:
        return Z, len(event_x), event_x, event_y, event_is_clustered
    return Z, len(event_x)

# Generate one displayed realization. Additional independent realizations are
# generated below when estimating correlation-function error bands.
Z, N_events, event_x, event_y, event_is_clustered = generate_synthetic_galaxy(rng, return_events=True)
print(
    f"Expected number of events: {expected_total:.0f}, realization 1: {N_events} "
    f"({np.sum(event_is_clustered)} clustered)"
)

# Visualisation of event distribution for the displayed realization.
bins = 500
fig, axes = plt.subplots(1, 2, figsize=(11, 5), sharex=True, sharey=True)
h0 = axes[0].hist2d(event_x, event_y, bins=bins, cmap='inferno')
plt.colorbar(h0[3], ax=axes[0], label='counts')
axes[0].set_title('All enrichment sources')
if np.any(event_is_clustered):
    h1 = axes[1].hist2d(event_x[event_is_clustered], event_y[event_is_clustered], bins=bins, cmap='magma')
    plt.colorbar(h1[3], ax=axes[1], label='clustered counts')
else:
    axes[1].text(0.5, 0.5, 'clustering disabled', transform=axes[1].transAxes,
                 ha='center', va='center')
axes[1].set_title('Clustered component')
for ax in axes:
    ax.set_xlim(-R_out, R_out)
    ax.set_ylim(-R_out, R_out)
    ax.set_aspect('equal')
    ax.set_xlabel('x (kpc)')
axes[0].set_ylabel('y (kpc)')
fig.suptitle(
    'Conditionally Poisson source density'
    if not enable_source_clustering else
    f'Thomas extension: f_cl={cluster_fraction}, '
    f'sigma_cl={cluster_radius} kpc'
)
plt.tight_layout()
plt.show()
```

```python
# %% Block 4: Subtraction of the deterministic axisymmetric mean
R_grid = np.sqrt(X**2 + Y**2)
disk_mask = (R_grid >= R_in) & (R_grid <= R_out)
# Use the same slightly eroded analysis window as the analytic finite-window
# calculation, keeping every pixel pair endpoint away from the source edges.
window_margin = np.sqrt(2.0) * dx
analysis_R_in = R_in + window_margin
analysis_R_out = R_out - window_margin

def subtract_axisymmetric_mean(Z, make_plot=False):
    """Subtract the deterministic local-WKBJ axisymmetric ensemble mean."""
    Z_bg = np.full_like(Z, np.nan, dtype=float)
    Z_bg[disk_mask] = (
        Y_val * lambda0_0 * t_star
        * np.exp(-R_grid[disk_mask] / h_R)
    )

    Z_res = np.full_like(Z, np.nan, dtype=float)
    Z_res[disk_mask] = Z[disk_mask] - Z_bg[disk_mask]
    Z_plot = np.where(disk_mask, Z, np.nan)
    Z_res_plot = np.where(disk_mask, Z_res, np.nan)

    if make_plot:
        fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))
        Z_bg_plot = np.where(disk_mask, Z_bg, np.nan)
        abundance_vmin, abundance_vmax = np.nanpercentile(Z_plot, [1, 99])

        im_abundance = axes[0].pcolormesh(
            X, Y, Z_plot, cmap='plasma', shading='auto',
            vmin=abundance_vmin, vmax=abundance_vmax
        )
        plt.colorbar(im_abundance, ax=axes[0], label=r'$\Sigma_X$')
        axes[0].set_title('Abundance')

        im_gradient = axes[1].pcolormesh(
            X, Y, Z_bg_plot, cmap='plasma', shading='auto',
            vmin=abundance_vmin, vmax=abundance_vmax
        )
        plt.colorbar(im_gradient, ax=axes[1], label=r'$\bar{\mu}(\varpi)$')
        axes[1].set_title('Axisymmetric ensemble mean')

        vmax = np.nanmax(np.abs(Z_res_plot))
        im_residual = axes[2].pcolormesh(
            X, Y, Z_res_plot, cmap='RdBu_r', shading='auto',
            vmin=-vmax, vmax=vmax
        )
        plt.colorbar(im_residual, ax=axes[2], label=r'$\delta\Sigma_{X,{\rm ax}}$')
        axes[2].set_title('Axisymmetric residual')

        for ax in axes:
            ax.set_xlim(-R_out, R_out)
            ax.set_ylim(-R_out, R_out)
            ax.set_aspect('equal')
            ax.set_xlabel(r'$x\;({\rm kpc})$')
        axes[0].set_ylabel(r'$y\;({\rm kpc})$')

        plt.tight_layout()
        fig.savefig(
            validation_figure_path('scalar_fields'), dpi=160,
            bbox_inches='tight'
        )
        plt.show()

    return {
        'Z_bg': Z_bg,
        'Z_res': Z_res,
        'Z_plot': Z_plot,
        'Z_res_plot': Z_res_plot,
    }

first_residual = subtract_axisymmetric_mean(Z, make_plot=True)
Z_bg = first_residual['Z_bg']
Z_res = first_residual['Z_res']
Z_plot = first_residual['Z_plot']
Z_res_plot = first_residual['Z_res_plot']

```

```python
# %% Block 5: Two-point statistics measured from the simulated maps
# Test midpoint radii are kept inside the eroded observing window. Each
# reported radius represents a midpoint annulus one pixel wide.
radii_test = np.round(np.linspace(analysis_R_in + 0.5, analysis_R_out - 0.5, 6), 2)
midpoint_annulus_halfwidth = 0.5 * dx

# Pair separations are restricted to the physical outer radius of the disc.
max_pair_separation = R_out
r_bins = np.linspace(0.1, max_pair_separation, 30)
r_centers = 0.5*(r_bins[:-1] + r_bins[1:])
sector_last_bin = np.flatnonzero(
    r_centers <= sector_validation_rmax
)[-1]
full_pair_block_size = 64

def measure_full_pair_statistics(Z_res):
    """Measure xi2 and xi3 from every pixel pair in the analysis window."""
    pixel_mask = (
        np.isfinite(Z_res)
        & (R_grid >= analysis_R_in)
        & (R_grid <= analysis_R_out)
    )
    pair_x = X[pixel_mask]
    pair_y = Y[pixel_mask]
    pair_value = Z_res[pixel_mask]
    n_pixel = pair_value.size
    n_separation_bin = len(r_centers)
    n_midpoint_bin = len(radii_test)

    xi3_product_sum = np.zeros(n_separation_bin)
    xi3_power_sum = np.zeros(n_separation_bin)
    xi3_pair_count = np.zeros(n_separation_bin, dtype=np.int64)
    sector_product_sum = np.zeros(n_separation_bin)
    sector_power_sum = np.zeros(n_separation_bin)
    sector_pair_count = np.zeros(n_separation_bin, dtype=np.int64)
    xi2_product_sum = np.zeros((n_midpoint_bin, n_separation_bin))
    xi2_power_sum = np.zeros((n_midpoint_bin, n_separation_bin))
    xi2_pair_count = np.zeros(
        (n_midpoint_bin, n_separation_bin), dtype=np.int64
    )

    # Enumerate only i < j. For the corresponding complete ordered-pair
    # sample, each endpoint appears in both orders, so its two variance sums
    # are identical and equal to the mean of z_i^2 and z_j^2 below.
    all_index = np.arange(n_pixel)[None, :]
    all_value = pair_value[None, :]
    all_power = all_value**2
    print(
        f'Exhaustive pair accumulation over {n_pixel:,} analysis pixels '
        f'({n_pixel * (n_pixel - 1) // 2:,} unordered pairs before cuts)'
    )

    for start in range(0, n_pixel, full_pair_block_size):
        stop = min(start + full_pair_block_size, n_pixel)
        block_index = np.arange(start, stop)[:, None]
        upper_triangle = all_index > block_index

        delta_x = pair_x[None, :] - pair_x[start:stop, None]
        delta_y = pair_y[None, :] - pair_y[start:stop, None]
        separation = np.hypot(delta_x, delta_y)
        separation_bin = (
            np.searchsorted(r_bins, separation, side='right') - 1
        )
        valid_pair = (
            upper_triangle
            & (separation_bin >= 0)
            & (separation_bin < n_separation_bin)
        )

        block_value = pair_value[start:stop, None]
        pair_product = block_value * all_value
        symmetric_endpoint_power = 0.5 * (block_value**2 + all_power)

        midpoint_x = 0.5 * (
            pair_x[None, :] + pair_x[start:stop, None]
        )
        midpoint_y = 0.5 * (
            pair_y[None, :] + pair_y[start:stop, None]
        )
        midpoint_radius = np.hypot(midpoint_x, midpoint_y)
        full_window_selected = (
            valid_pair
            & (midpoint_radius >= analysis_R_in)
            & (midpoint_radius <= analysis_R_out)
        )
        selected_bin = separation_bin[full_window_selected]
        xi3_pair_count += np.bincount(
            selected_bin, minlength=n_separation_bin
        )
        xi3_product_sum += np.bincount(
            selected_bin, weights=pair_product[full_window_selected],
            minlength=n_separation_bin
        )
        xi3_power_sum += np.bincount(
            selected_bin, weights=symmetric_endpoint_power[full_window_selected],
            minlength=n_separation_bin
        )
        # The phase-sector diagnostic is reported only to sector_validation_rmax.
        # Compute phases after the geometric cuts and only for those bins.
        sector_candidate = (
            full_window_selected
            & (midpoint_radius >= sector_analysis_R_in)
            & (midpoint_radius <= sector_analysis_R_out)
            & (separation_bin <= sector_last_bin)
        )
        candidate_phi = np.arctan2(
            midpoint_y[sector_candidate],
            midpoint_x[sector_candidate]
        )
        candidate_theta = spiral_phase(
            midpoint_radius[sector_candidate], candidate_phi
        )
        wrapped_theta = (
            (candidate_theta + np.pi) % (2*np.pi) - np.pi
        )
        in_arm_phase_sector = (
            np.abs(wrapped_theta) < sector_phase_halfwidth
        )
        sector_bin = separation_bin[sector_candidate][
            in_arm_phase_sector
        ]
        sector_product = pair_product[sector_candidate][
            in_arm_phase_sector
        ]
        sector_power = symmetric_endpoint_power[sector_candidate][
            in_arm_phase_sector
        ]
        sector_pair_count += np.bincount(
            sector_bin, minlength=n_separation_bin
        )
        sector_product_sum += np.bincount(
            sector_bin, weights=sector_product,
            minlength=n_separation_bin
        )
        sector_power_sum += np.bincount(
            sector_bin, weights=sector_power,
            minlength=n_separation_bin
        )
        for radius_index, varpi_m in enumerate(radii_test):
            in_midpoint_annulus = (
                np.abs(midpoint_radius - varpi_m)
                < midpoint_annulus_halfwidth
            )
            selected = valid_pair & in_midpoint_annulus
            selected_bin = separation_bin[selected]
            xi2_pair_count[radius_index] += np.bincount(
                selected_bin, minlength=n_separation_bin
            )
            xi2_product_sum[radius_index] += np.bincount(
                selected_bin, weights=pair_product[selected],
                minlength=n_separation_bin
            )
            xi2_power_sum[radius_index] += np.bincount(
                selected_bin, weights=symmetric_endpoint_power[selected],
                minlength=n_separation_bin
            )

    xi3 = np.divide(
        xi3_product_sum, xi3_power_sum,
        out=np.full(n_separation_bin, np.nan),
        where=(xi3_pair_count > 0) & (xi3_power_sum > 0),
    )
    xi2_array = np.divide(
        xi2_product_sum, xi2_power_sum,
        out=np.full_like(xi2_product_sum, np.nan),
        where=(xi2_pair_count > 0) & (xi2_power_sum > 0),
    )
    xi_sector = np.divide(
        sector_product_sum, sector_power_sum,
        out=np.full(n_separation_bin, np.nan),
        where=(sector_pair_count > 0) & (sector_power_sum > 0),
    )
    return {
        'xi2': {
            varpi_m: xi2_array[index]
            for index, varpi_m in enumerate(radii_test)
        },
        'xi3': xi3,
        'xi_sector': xi_sector,
        'xi2_pair_count': xi2_pair_count,
        'xi3_pair_count': xi3_pair_count,
        'sector_pair_count': sector_pair_count,
    }

# Collect independent numerical realizations for error bands.
first_pair_statistics = measure_full_pair_statistics(Z_res)
xi2_realizations = [first_pair_statistics['xi2']]
xi3_realizations = [first_pair_statistics['xi3']]
sector_realizations = [first_pair_statistics['xi_sector']]
for ireal in range(1, n_realizations):
    Z_i, N_i = generate_synthetic_galaxy(rng, return_events=False)
    res_i = subtract_axisymmetric_mean(Z_i, make_plot=False)
    pair_statistics_i = measure_full_pair_statistics(res_i['Z_res'])
    xi2_realizations.append(pair_statistics_i['xi2'])
    xi3_realizations.append(pair_statistics_i['xi3'])
    sector_realizations.append(pair_statistics_i['xi_sector'])
    print(
        f"Completed realization {ireal+1}/{n_realizations}: "
        f"N_events={N_i}, accepted full-window pairs="
        f"{pair_statistics_i['xi3_pair_count'].sum():,}"
    )

sim_corr_mean = {
    varpi_m: np.nanmean(np.vstack([real[varpi_m] for real in xi2_realizations]), axis=0)
    for varpi_m in radii_test
}
sim_corr_std = {
    varpi_m: np.nanstd(np.vstack([real[varpi_m] for real in xi2_realizations]), axis=0, ddof=1)
    if n_realizations > 1 else np.zeros_like(sim_corr_mean[varpi_m])
    for varpi_m in radii_test
}
sim_corr_sem = {
    varpi_m: sim_corr_std[varpi_m] / np.sqrt(n_realizations)
    for varpi_m in radii_test
}
sector_stack = np.vstack(sector_realizations)
sector_sim_mean = np.nanmean(sector_stack, axis=0)
sector_sim_std = np.nanstd(sector_stack, axis=0, ddof=1)
sector_sim_sem = sector_sim_std / np.sqrt(n_realizations)
```

```python
# %% Block 6: KTZ prediction
from functools import lru_cache

p0_val = sigma_scalar**2
p_ast_val = sigma_scalar**2 + 2*kappa*t_star
component_order = ['CP0', 'CPN', 'CM']
if enable_source_clustering:
    component_order.append('CNP')
component_colors = {
    'CP0': '#7fc97f',
    'CPN': '#ffff99',
    'CM': '#fdc086',
    'CNP': '#beaed4',
}


component_labels = {
    'CP0': r'$C_{\rm 2p,0}$',
    'CPN': r'$C_{\rm 2p,N}$',
    'CM': r'$C_{\rm M}$',
    'CNP': r'$C_{\rm NP}$',
}

def _square_pixel_power_average_scalar(Q, n_angle=128):
    """Orientation-averaged squared Fourier response of a square pixel."""
    angle = np.linspace(0.0, 2 * np.pi, n_angle, endpoint=False)
    kx = Q * np.cos(angle)
    ky = Q * np.sin(angle)
    window = (
        np.sinc(kx * dx / (2 * np.pi))
        * np.sinc(ky * dx / (2 * np.pi))
    )
    return float(np.mean(window**2))


# Fixed scrambled Sobol samples integrate the exact square-pixel difference
# distribution and the orientation of the separation relative to the fixed pixel axes.
_scalar_pixel_qmc = qmc.Sobol(d=5, scramble=True, seed=20260929)
_scalar_pixel_sample = _scalar_pixel_qmc.random_base2(
    int(np.log2(scalar_pixel_response_samples))
)
_scalar_delta_x = dx * (
    _scalar_pixel_sample[:, 2] - _scalar_pixel_sample[:, 0]
)
_scalar_delta_y = dx * (
    _scalar_pixel_sample[:, 3] - _scalar_pixel_sample[:, 1]
)
_scalar_pixel_angle = 2 * np.pi * _scalar_pixel_sample[:, 4]
_scalar_pixel_cos = np.cos(_scalar_pixel_angle)
_scalar_pixel_sin = np.sin(_scalar_pixel_angle)


def _pixel_shifted_radius(r):
    return np.hypot(
        r * _scalar_pixel_cos + _scalar_delta_x,
        r * _scalar_pixel_sin + _scalar_delta_y,
    )


def _I0_continuous(radius):
    radius = np.asarray(radius, dtype=float)
    radius2 = radius**2
    result = 0.5 * (
        exp1(radius2 / (4 * p_ast_val))
        - exp1(radius2 / (4 * p0_val))
    )
    at_zero = radius == 0
    if np.any(at_zero):
        result = np.asarray(result)
        result[at_zero] = 0.5 * np.log(p_ast_val / p0_val)
    return result


_scalar_inp_radius = np.linspace(
    0.0, max_pair_separation + np.sqrt(2.0) * dx, 8193
)
_scalar_inp_quad_x, _scalar_inp_quad_w = np.polynomial.legendre.leggauss(48)
_scalar_inp_p = (
    0.5 * (p_ast_val - p0_val) * _scalar_inp_quad_x
    + 0.5 * (p_ast_val + p0_val)
)
_scalar_inp_weight = 0.5 * (p_ast_val - p0_val) * _scalar_inp_quad_w


@lru_cache(maxsize=None)
def _InP_continuous_table(Qn_key):
    Qn = float(Qn_key)
    radius2 = _scalar_inp_radius[:, None]**2
    integrand = (
        0.5 / _scalar_inp_p[None, :]
        * np.exp(
            -radius2 / (4 * _scalar_inp_p[None, :])
            - Qn**2 * _scalar_inp_p[None, :] / 4
        )
    )
    return integrand @ _scalar_inp_weight


@lru_cache(maxsize=None)
def I0(r):
    """Same-event kernel with the exact orientation-averaged square response."""
    return float(np.mean(_I0_continuous(_pixel_shifted_radius(float(r)))))


@lru_cache(maxsize=None)
def InP(r, Qn):
    """Spiral same-event kernel with the matched square-pixel response."""
    Qn_key = round(float(Qn), 12)
    continuous = np.interp(
        _pixel_shifted_radius(float(r)),
        _scalar_inp_radius,
        _InP_continuous_table(Qn_key),
    )
    # The local-WKBJ covariance places the spiral phase at the pair midpoint.
    # Averaging both square pixels damps this midpoint phase by W(Q/2)^2.
    midpoint_window = _square_pixel_power_average_scalar(0.5 * Qn)
    return float(np.mean(continuous) * midpoint_window)


@lru_cache(maxsize=None)
def Tn(Qn):
    if Qn == 0:
        return t_star
    continuous = (
        np.exp(-0.5 * sigma_scalar**2 * Qn**2)
        * (1 - np.exp(-kappa * t_star * Qn**2))
        / (kappa * Qn**2)
    )
    # Each mean field is averaged over one square pixel. The square root is
    # used because the covariance contains the product of two such means.
    return continuous * np.sqrt(
        _square_pixel_power_average_scalar(Qn)
    )

# Thomas-process source clustering adds a non-Poisson covariance term.  We model
# the source pair correlation as Gaussian with correlation length
# l_src = sqrt(2)*cluster_radius, which corresponds to child offsets drawn with
# one-dimensional dispersion cluster_radius around each parent complex.
_cluster_quad_x, _cluster_quad_w = np.polynomial.legendre.leggauss(cluster_theory_nodes)
_cluster_t_nodes = 0.5*t_star*(_cluster_quad_x + 1.0)
_cluster_t_weights = 0.5*t_star*_cluster_quad_w
_cluster_p_nodes = sigma_scalar**2 + 2*kappa*_cluster_t_nodes
_cluster_source_corr_len2 = 2.0*cluster_radius**2

_cluster_denom = (
    _cluster_source_corr_len2
    + _cluster_p_nodes[:, None]
    + _cluster_p_nodes[None, :]
)
_cluster_weighted_prefactor = (
    _cluster_t_weights[:, None] * _cluster_t_weights[None, :]
    * _cluster_source_corr_len2 / _cluster_denom
)


@lru_cache(maxsize=1)
def _I_cluster_np_continuous_table():
    """Continuous C_NP kernel tabulated before pixel averaging."""
    result = np.empty_like(_scalar_inp_radius)
    chunk_size = 256
    for start in range(0, len(_scalar_inp_radius), chunk_size):
        stop = min(start + chunk_size, len(_scalar_inp_radius))
        radius2 = _scalar_inp_radius[start:stop, None, None]**2
        result[start:stop] = np.sum(
            _cluster_weighted_prefactor[None, :, :]
            * np.exp(-radius2 / (2.0 * _cluster_denom[None, :, :])),
            axis=(1, 2),
        )
    return result


@lru_cache(maxsize=None)
def I_cluster_np(r):
    """Isotropic C_NP kernel with the matched square-pixel response."""
    if (not enable_source_clustering) or cluster_fraction <= 0 or cluster_radius <= 0:
        return 0.0
    continuous = np.interp(
        _pixel_shifted_radius(float(r)),
        _scalar_inp_radius,
        _I_cluster_np_continuous_table(),
    )
    return float(np.mean(continuous))

def CNP_prefactor(varpi_m):
    if (not enable_source_clustering) or cluster_fraction <= 0 or cluster_radius <= 0:
        return 0.0
    lam_local = lambda0_0 * np.exp(-varpi_m/h_R)
    lam_clustered = max(cluster_fraction * lam_local, 1.0e-30)
    # Parent complexes are sampled over the full age window.  The Thomas-process
    # pair-correlation amplitude therefore uses the integrated clustered surface
    # density Sigma_cl = lambda_clustered*t_star, not the instantaneous rate.
    sigma_clustered = lam_clustered * t_star
    xi0 = mean_children_per_cluster / (4*np.pi*sigma_clustered*cluster_radius**2)
    # The total source covariance scales as lambda_total * cluster_fraction^2 * xi_src.
    return 4*np.pi*kappa*lam_local*(cluster_fraction**2)*xi0

def A_2p_factor(varpi_m):
    r"""Common physical covariance scale $A_{\rm 2p}(\varpi)$."""
    lam_local = lambda0_0 * np.exp(-varpi_m/h_R)
    return (Y_val**2) * lam_local / (4*np.pi*kappa)

def source_harmonic_damping(Qn):
    """Damping of the source-rate spiral harmonic caused by clustered child offsets."""
    if (not enable_source_clustering) or cluster_fraction <= 0 or cluster_radius <= 0:
        return 1.0
    f_cl = np.clip(cluster_fraction, 0.0, 1.0)
    child_damping = np.exp(-0.5*(Qn*cluster_radius)**2)
    return (1.0 - f_cl) + f_cl*child_damping

def local_cov_component_parts(varpi_m, phi_m, r, psi):
    Theta_m = spiral_phase(varpi_m, phi_m)
    Q_base = k_perp(varpi_m)
    dPhi = (m_arms/np.tan(pitch_rad))/varpi_m
    grad_x = dPhi*np.cos(phi_m) + (m_arms/varpi_m)*np.sin(phi_m)
    grad_y = dPhi*np.sin(phi_m) - (m_arms/varpi_m)*np.cos(phi_m)
    grad = np.array([grad_x, grad_y])
    r_vec = r*np.array([np.cos(psi), np.sin(psi)])

    # Linearized endpoint phases: Theta_+ - Theta_- = grad(Theta) dot r_vec.
    dot = np.dot(grad, r_vec)
    Theta_minus = Theta_m - 0.5*dot
    Theta_plus  = Theta_m + 0.5*dot

    lam_local = lambda0_0 * np.exp(-varpi_m/h_R)
    R_val = 4*np.pi*kappa*lam_local*eta**2

    CPN_num = 0.0
    CPN_var_m = 0.0
    CPN_var_p = 0.0
    mean_minus = 0.0
    mean_plus = 0.0

    for n, g, alpha in zip(harmonic_n, harmonic_g, harmonic_alpha):
        Qn = n * Q_base
        damping = source_harmonic_damping(Qn)
        S_m = g*damping*np.cos(n*Theta_m + alpha)
        S_mi = g*damping*np.cos(n*Theta_minus + alpha)
        S_pl = g*damping*np.cos(n*Theta_plus + alpha)

        CPN_num += InP(r, Qn)*S_m
        CPN_var_m += InP(0, Qn)*S_mi
        CPN_var_p += InP(0, Qn)*S_pl

        Tn_val = Tn(Qn)
        mean_minus += Tn_val*S_mi
        mean_plus += Tn_val*S_pl

    # Numerator covariance pieces, after extracting the common A_2p factor.
    nums = {
        'CP0': I0(r),
        'CPN': eta*CPN_num,
        'CM': R_val * mean_minus * mean_plus,
        'CNP': CNP_prefactor(varpi_m) * I_cluster_np(r),
    }
    # Endpoint variance pieces. CNP uses the same local clustering approximation.
    vars_m_parts = {
        'CP0': I0(0),
        'CPN': eta*CPN_var_m,
        'CM': R_val*mean_minus**2,
        'CNP': CNP_prefactor(varpi_m) * I_cluster_np(0.0),
    }
    vars_p_parts = {
        'CP0': I0(0),
        'CPN': eta*CPN_var_p,
        'CM': R_val*mean_plus**2,
        'CNP': CNP_prefactor(varpi_m) * I_cluster_np(0.0),
    }
    return nums, vars_m_parts, vars_p_parts

def endpoint_window_ok(varpi_m, phi_m, r, psi):
    if r < 0 or r > max_pair_separation:
        return False
    mid_x = varpi_m*np.cos(phi_m)
    mid_y = varpi_m*np.sin(phi_m)
    hx = 0.5*r*np.cos(psi)
    hy = 0.5*r*np.sin(psi)
    rm = np.hypot(mid_x - hx, mid_y - hy)
    rp = np.hypot(mid_x + hx, mid_y + hy)
    return (analysis_R_in <= rm <= analysis_R_out) and (analysis_R_in <= rp <= analysis_R_out)

def _component_average_from_samples(sample_parts, sample_weights=None):
    if not sample_parts:
        return {name: np.nan for name in component_order}

    if sample_weights is None:
        sample_weights = np.ones(len(sample_parts), dtype=float)
    else:
        sample_weights = np.asarray(sample_weights, dtype=float)
        if len(sample_weights) != len(sample_parts):
            raise ValueError("sample_weights must match sample_parts")

    nums_by_component = {name: [] for name in component_order}
    vars_m_total, vars_p_total = [], []
    for weight, (nums, vars_m_parts, vars_p_parts) in zip(sample_weights, sample_parts):
        if (not np.isfinite(weight)) or weight <= 0:
            continue
        var_m = sum(vars_m_parts.values())
        var_p = sum(vars_p_parts.values())
        if var_m <= 0 or var_p <= 0:
            continue
        # The local expressions have the common scale A_2p factored out.
        # For fixed-varpi xi2 this cancels, but full-window xi3 must average the
        # physical covariance and variance before normalization.
        weighted_var_m = weight * var_m
        weighted_var_p = weight * var_p
        for name in component_order:
            nums_by_component[name].append(weight * nums[name])
        vars_m_total.append(weighted_var_m)
        vars_p_total.append(weighted_var_p)

    if not vars_m_total:
        return {name: np.nan for name in component_order}

    denom = np.sqrt(np.mean(vars_m_total)*np.mean(vars_p_total))
    return {
        name: np.mean(nums_by_component[name]) / denom
        for name in component_order
    }

def xi2_KTZ_components(varpi_m, r, Nphi=40, Npsi=40):
    phi_arr = np.linspace(0,2*np.pi,Nphi,endpoint=False)
    psi_arr = np.linspace(0,2*np.pi,Npsi,endpoint=False)
    sample_parts = []
    for phi in phi_arr:
        for psi in psi_arr:
            if endpoint_window_ok(varpi_m, phi, r, psi):
                sample_parts.append(local_cov_component_parts(varpi_m, phi, r, psi))
    return _component_average_from_samples(sample_parts)

pair_bin_quadrature_nodes = 5
_pair_quad_x, _pair_quad_w = np.polynomial.legendre.leggauss(
    pair_bin_quadrature_nodes
)

def separation_bin_quadrature(r_lower, r_upper):
    """Return separation nodes and the two-dimensional pair measure r dr."""
    radius = (
        0.5 * (r_upper - r_lower) * _pair_quad_x
        + 0.5 * (r_upper + r_lower)
    )
    weight = (
        0.5 * (r_upper - r_lower) * _pair_quad_w * radius
    )
    return radius, weight

def xi2_KTZ_bin_components(
    varpi_m, r_lower, r_upper, Nphi=40, Npsi=40
):
    """Average physical pair products over a finite separation bin."""
    phi_arr = np.linspace(0, 2*np.pi, Nphi, endpoint=False)
    psi_arr = np.linspace(0, 2*np.pi, Npsi, endpoint=False)
    radius_arr, radius_weight = separation_bin_quadrature(
        r_lower, r_upper
    )
    sample_parts = []
    sample_weights = []
    for radius, weight in zip(radius_arr, radius_weight):
        for phi in phi_arr:
            for psi in psi_arr:
                if endpoint_window_ok(varpi_m, phi, radius, psi):
                    sample_parts.append(local_cov_component_parts(
                        varpi_m, phi, radius, psi
                    ))
                    sample_weights.append(weight)
    return _component_average_from_samples(
        sample_parts, sample_weights
    )

def components_to_arrays(component_list):
    return {name: np.array([comp[name] for comp in component_list]) for name in component_order}


def plot_theory_component_stack(
    ax, r_plot, comp_plot, alpha=0.42, labels=None
):
    if labels is None:
        labels = component_labels
    cumulative = np.zeros_like(r_plot, dtype=float)
    handles = []
    for name in component_order:
        vals = comp_plot[name]
        next_cumulative = cumulative + vals
        handle = ax.fill_between(
            r_plot, cumulative, next_cumulative,
            color=component_colors[name], alpha=alpha, linewidth=0,
            label=labels[name], zorder=1,
        )
        handles.append(handle)
        cumulative = next_cumulative
    return handles

wkb_components = {}
for varpi_m in radii_test:
    comp_list = [
        xi2_KTZ_bin_components(varpi_m, r_lower, r_upper)
        for r_lower, r_upper in zip(r_bins[:-1], r_bins[1:])
    ]
    wkb_components[varpi_m] = components_to_arrays(comp_list)

# Full azimuthal averaging in the circular annulus cancels every linear
# harmonic at the covariance level. The matched square-pixel response is
# orientation averaged, so it preserves this cancellation exactly.
xi2_CPN_max = max(
    np.nanmax(np.abs(wkb_components[varpi_m]['CPN']))
    for varpi_m in radii_test
)
assert xi2_CPN_max < 1.0e-12
```

```python
# %% Block 6b: Local xi1 diagnostic before annular averaging
# This diagnostic fixes the midpoint radius and spiral phase, so the signed CPN
# contribution is not averaged away over a full annulus. It is useful for
# checking the local theory before comparing xi2/xi3.

xi1_varpi_test = 7.5
xi1_phase_cases = {
    'arm ridge': 0.0,
    'inter-arm': np.pi,
    'quadrature': 0.5*np.pi,
}
xi1_psi_local = 0.0  # separation along the local arm-normal direction
xi1_r = r_centers

def phi_for_theta(varpi, theta_target):
    """Return a disc azimuth whose midpoint has the requested spiral phase."""
    Phi = (m_arms / np.tan(pitch_rad)) * np.log(varpi / R_in)
    return (Phi + Theta0 - theta_target) / m_arms

def xi1_components(varpi, theta_target, r, psi_local=0.0):
    phi_m = phi_for_theta(varpi, theta_target)
    dPhi = (m_arms/np.tan(pitch_rad))/varpi
    grad_x = dPhi*np.cos(phi_m) + (m_arms/varpi)*np.sin(phi_m)
    grad_y = dPhi*np.sin(phi_m) - (m_arms/varpi)*np.cos(phi_m)
    grad_angle = np.arctan2(grad_y, grad_x)
    psi_global = grad_angle + psi_local
    nums, var_m_parts, var_p_parts = local_cov_component_parts(varpi, phi_m, r, psi_global)
    var_m = sum(var_m_parts.values())
    var_p = sum(var_p_parts.values())
    if var_m <= 0 or var_p <= 0:
        return {name: np.nan for name in component_order}
    denom = np.sqrt(var_m*var_p)
    return {name: nums[name]/denom for name in component_order}

fig, axes = plt.subplots(1, len(xi1_phase_cases), figsize=(15, 4), sharey=True)
if len(xi1_phase_cases) == 1:
    axes = [axes]

for ax, (label, theta_target) in zip(axes, xi1_phase_cases.items()):
    comp_list = [xi1_components(xi1_varpi_test, theta_target, r, xi1_psi_local) for r in xi1_r]
    comp_arrays = components_to_arrays(comp_list)
    comp0 = xi1_components(xi1_varpi_test, theta_target, 0.0, xi1_psi_local)
    comp_plot = {
        name: np.concatenate([[comp0[name]], comp_arrays[name]])
        for name in component_order
    }
    r_plot = np.concatenate([[0.0], xi1_r])
    xi1_total = sum(comp_plot[name] for name in component_order)

    plot_theory_component_stack(ax, r_plot, comp_plot, alpha=0.5)
    ax.plot(r_plot, xi1_total, 'k--', lw=1.5, label=r'$\xi_1$ total')
    ax.axhline(0, color='grey', lw=0.5)
    ax.set_title(fr'{label}, $\Theta={theta_target:.2f}$')
    ax.set_xlabel('r (kpc)')
    ax.set_xlim(left=0)
    ax.grid(alpha=0.25)
    ax.legend(fontsize=8, ncol=2)

axes[0].set_ylabel(r'$\xi_1(\boldsymbol{x}_m, r)$')
fig.suptitle(fr'Local $\xi_1$ decomposition at $\varpi_m={xi1_varpi_test}$ kpc')
plt.tight_layout()
plt.show()
```

```python
# %% Block 7a: Comparison of correlation functions
correlation_plot_rmax = validation_rmax
annular_component_labels = component_labels.copy()
annular_component_labels['CPN'] = r'$C_{\rm 2p,N}=0$'

fig, axes = plt.subplots(2, 3, figsize=(15, 9), sharey=True)
axes = axes.flatten()

for panel_index, (ax, varpi_m) in enumerate(zip(axes, radii_test)):
    r_plot = np.concatenate([[0.0], r_centers])
    sim_mean = np.concatenate([[1.0], sim_corr_mean[varpi_m]])
    sim_std = np.concatenate([[0.0], sim_corr_std[varpi_m]])
    sim_sem = np.concatenate([[0.0], sim_corr_sem[varpi_m]])
    comp0 = xi2_KTZ_components(varpi_m, 0.0)
    comp_plot = {
        name: np.concatenate([[comp0[name]], wkb_components[varpi_m][name]])
        for name in component_order
    }
    wkb_plot = sum(comp_plot[name] for name in component_order)

    plot_theory_component_stack(
        ax, r_plot, comp_plot, labels=annular_component_labels
    )
    ax.plot(r_plot, sim_mean, 'o-', color='black', ms=4, label='Simulation mean', zorder=4)
    ax.fill_between(r_plot, sim_mean-sim_std, sim_mean+sim_std,
                    color='black', alpha=0.16, linewidth=0,
                    label=r'Single-realisation $1\sigma$', zorder=3)
    ax.errorbar(
        r_plot, sim_mean, yerr=sim_sem, fmt='none',
        ecolor='black', elinewidth=0.9, capsize=2,
        label=r'Mean $1\sigma$ uncertainty', zorder=5
    )
    ax.plot(r_plot, wkb_plot, 'k--', lw=1.5, label='KTZ total', zorder=5)
    ax.axhline(0, color='grey', lw=0.5)
    ax.set_title(fr'$\varpi = {varpi_m}$ kpc')
    ax.set_xlabel('r (kpc)')
    ax.set_xlim(0, correlation_plot_rmax)
    if panel_index == 0:
        ax.legend(loc='lower left', fontsize=8, ncol=2)

axes[0].set_ylabel(r'$\xi_{2,{\rm pp}}(\varpi,r)$')
plt.tight_layout()
fig.savefig(
    validation_figure_path('scalar_xi2'), dpi=160,
    bbox_inches='tight'
)
plt.show()

# %% Block 7b: Residual
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
axes = axes.flatten()

for i, (ax, varpi_m) in enumerate(zip(axes, radii_test)):
    r_plot = np.concatenate([[0.0], r_centers])
    comp0 = xi2_KTZ_components(varpi_m, 0.0)
    wkb_plot = sum(
        np.concatenate([[comp0[name]], wkb_components[varpi_m][name]])
        for name in component_order
    )
    diff = np.concatenate([[1.0], sim_corr_mean[varpi_m]]) - wkb_plot
    diff[0] = 0.0
    sim_sem = np.concatenate([[0.0], sim_corr_sem[varpi_m]])
    ax.plot(
        r_plot, diff, 'o-',
        label=r'Simulation mean $-$ KTZ ($C_{\rm 2p,N}=0$)'
    )
    ax.fill_between(
        r_plot, diff - sim_sem, diff + sim_sem,
        color='C0', alpha=0.22, linewidth=0,
        label=r'Mean $1\sigma$ uncertainty'
    )
    ax.axhline(0, color='grey', linestyle='--', linewidth=0.8)

    ax.set_title(fr'$\varpi = {varpi_m}$ kpc')
    ax.set_xlabel('r (kpc)')
    ax.set_ylabel(r'$\Delta\xi_2$ (Sim $-$ KTZ)')
    ax.set_xlim(0, correlation_plot_rmax)

    if i == 0:
        ax.legend(loc='lower left', fontsize=8)

fig.suptitle(
    r'Residual (Simulation mean $-$ KTZ) with $1\sigma$ mean uncertainty; '
    r'$C_{\rm 2p,N}=0$'
)
plt.tight_layout()
plt.show()

# Curve-level accuracy metrics exclude r=0, where xi=1 is imposed by
# normalization rather than independently estimated from the maps.
def summarize_curve_validation(simulation, theory, uncertainty):
    simulation = np.asarray(simulation)
    theory = np.asarray(theory)
    uncertainty = np.asarray(uncertainty)
    valid = (
        np.isfinite(simulation)
        & np.isfinite(theory)
        & np.isfinite(uncertainty)
    )
    residual = simulation[valid] - theory[valid]
    return {
        'n': int(np.sum(valid)),
        'rmse': float(np.sqrt(np.mean(residual**2))),
        'bias': float(np.mean(residual)),
        'pearson': float(np.corrcoef(
            simulation[valid], theory[valid]
        )[0, 1]),
        'max_abs': float(np.max(np.abs(residual))),
        'within_1sigma': int(np.sum(
            np.abs(residual) <= uncertainty[valid]
        )),
    }


xi2_metric_mask = r_centers <= correlation_plot_rmax
xi2_metric_simulation = np.concatenate([
    sim_corr_mean[varpi_m][xi2_metric_mask]
    for varpi_m in radii_test
])
xi2_metric_theory = np.concatenate([
    sum(
        wkb_components[varpi_m][name]
        for name in component_order
    )[xi2_metric_mask]
    for varpi_m in radii_test
])
xi2_metric_uncertainty = np.concatenate([
    sim_corr_sem[varpi_m][xi2_metric_mask]
    for varpi_m in radii_test
])
xi2_validation_metrics = summarize_curve_validation(
    xi2_metric_simulation,
    xi2_metric_theory,
    xi2_metric_uncertainty,
)
print(
    'xi2 validation '
    f'(N={xi2_validation_metrics["n"]}): '
    f'RMSE={xi2_validation_metrics["rmse"]:.3f}, '
    f'mean(simulation - KTZ)='
    f'{xi2_validation_metrics["bias"]:+.3f}, '
    f'Pearson r={xi2_validation_metrics["pearson"]:.3f}, '
    f'max|residual|={xi2_validation_metrics["max_abs"]:.3f}, '
    f'within mean 1 sigma='
    f'{xi2_validation_metrics["within_1sigma"]}/'
    f'{xi2_validation_metrics["n"]}'
)
```

```python
# %% Block 8: Full-window and arm-phase-sector statistics
# Both theory curves average physical pair products over the same finite
# separation-bin edges as the simulation, with the planar pair measure r dr.
# The arm-phase sector is a conditionally Poisson diagnostic of C_2p,N; the
# optional isotropic C_NP extension is validated with complete-azimuth
# averages only.
xi3_stack = np.vstack(xi3_realizations)
xi3_sim_mean = np.nanmean(xi3_stack, axis=0)
xi3_sim_std = np.nanstd(xi3_stack, axis=0, ddof=1)
xi3_sim_sem = xi3_sim_std / np.sqrt(n_realizations)
validation_bin_mask = r_centers <= correlation_plot_rmax
validation_r_centers = r_centers[validation_bin_mask]
validation_r_lower = r_bins[:-1][validation_bin_mask]
validation_r_upper = r_bins[1:][validation_bin_mask]
sector_bin_mask = r_centers <= sector_validation_rmax
sector_r_centers = r_centers[sector_bin_mask]
sector_r_lower = r_bins[:-1][sector_bin_mask]
sector_r_upper = r_bins[1:][sector_bin_mask]

def xi3_KTZ_components(
    r, phase_limited=False, Nvarpi=30, Nphi=60, Npsi=20
):
    # Equal spacing in R^2 gives equal-area midpoint samples.
    radial_lower = sector_analysis_R_in if phase_limited else analysis_R_in
    radial_upper = sector_analysis_R_out if phase_limited else analysis_R_out
    rho2_edges = np.linspace(
        radial_lower**2, radial_upper**2, Nvarpi + 1
    )
    varpi_arr = np.sqrt(
        0.5 * (rho2_edges[:-1] + rho2_edges[1:])
    )
    phi_arr = np.linspace(0, 2*np.pi, Nphi, endpoint=False)
    psi_arr = np.linspace(0, 2*np.pi, Npsi, endpoint=False)
    sample_parts = []
    sample_weights = []
    for varpi in varpi_arr:
        physical_weight = A_2p_factor(varpi)
        for phi in phi_arr:
            if phase_limited:
                theta = spiral_phase(varpi, phi)
                if abs(np.angle(np.exp(1j * theta))) >= sector_phase_halfwidth:
                    continue
            for psi in psi_arr:
                if endpoint_window_ok(varpi, phi, r, psi):
                    sample_parts.append(local_cov_component_parts(
                        varpi, phi, r, psi
                    ))
                    sample_weights.append(physical_weight)
    return _component_average_from_samples(
        sample_parts, sample_weights=sample_weights
    )

def xi3_KTZ_bin_components(
    r_lower, r_upper, phase_limited=False,
    Nvarpi=30, Nphi=60, Npsi=20
):
    radius_arr, radius_weight = separation_bin_quadrature(
        r_lower, r_upper
    )
    radial_lower = sector_analysis_R_in if phase_limited else analysis_R_in
    radial_upper = sector_analysis_R_out if phase_limited else analysis_R_out
    rho2_edges = np.linspace(
        radial_lower**2, radial_upper**2, Nvarpi + 1
    )
    varpi_arr = np.sqrt(
        0.5 * (rho2_edges[:-1] + rho2_edges[1:])
    )
    phi_arr = np.linspace(0, 2*np.pi, Nphi, endpoint=False)
    psi_arr = np.linspace(0, 2*np.pi, Npsi, endpoint=False)
    sample_parts = []
    sample_weights = []
    for radius, radial_weight in zip(radius_arr, radius_weight):
        for varpi in varpi_arr:
            physical_weight = radial_weight * A_2p_factor(varpi)
            for phi in phi_arr:
                if phase_limited:
                    theta = spiral_phase(varpi, phi)
                    if abs(np.angle(np.exp(1j * theta))) >= sector_phase_halfwidth:
                        continue
                for psi in psi_arr:
                    if endpoint_window_ok(varpi, phi, radius, psi):
                        sample_parts.append(local_cov_component_parts(
                            varpi, phi, radius, psi
                        ))
                        sample_weights.append(physical_weight)
    return _component_average_from_samples(
        sample_parts, sample_weights=sample_weights
    )

xi3_component_list = [
    xi3_KTZ_bin_components(r_lower, r_upper)
    for r_lower, r_upper in zip(validation_r_lower, validation_r_upper)
]
xi3_ktz_components = components_to_arrays(xi3_component_list)
xi3_CPN_max = np.nanmax(np.abs(xi3_ktz_components['CPN']))
assert xi3_CPN_max < 1.0e-12
xi3_ktz = sum(xi3_ktz_components[name] for name in component_order)

r_plot = np.concatenate([[0.0], validation_r_centers])
sim_plot = np.concatenate([[1.0], xi3_sim_mean[validation_bin_mask]])
sim_std = np.concatenate([[0.0], xi3_sim_std[validation_bin_mask]])
sim_sem = np.concatenate([[0.0], xi3_sim_sem[validation_bin_mask]])
xi3_comp0 = xi3_KTZ_components(0.0)
xi3_comp_plot = {
    name: np.concatenate([[xi3_comp0[name]], xi3_ktz_components[name]])
    for name in component_order
}
ktz_plot = sum(xi3_comp_plot[name] for name in component_order)

fig, ax = plt.subplots(figsize=(7, 5))
plot_theory_component_stack(
    ax, r_plot, xi3_comp_plot, labels=annular_component_labels
)
ax.plot(
    r_plot, sim_plot, 'o-', color='black', ms=4,
    label=r'Simulation $\xi_{3,{\rm pp}}$', zorder=4
)
ax.fill_between(
    r_plot, sim_plot-sim_std, sim_plot+sim_std,
    color='black', alpha=0.14, linewidth=0,
    label=r'Single-realisation $1\sigma$', zorder=3
)
ax.errorbar(
    r_plot, sim_plot, yerr=sim_sem, fmt='none',
    ecolor='black', elinewidth=0.9, capsize=2,
    label=r'Mean $1\sigma$ uncertainty', zorder=5
)
ax.plot(
    r_plot, ktz_plot, 'k--', lw=1.5,
    label=r'KTZ $\xi_{3,{\rm pp}}$', zorder=6
)
ax.axhline(0, color='grey', lw=0.5)
ax.set_xlabel('r (kpc)')
ax.set_ylabel(r'$\xi_{3,{\rm pp}}(r)$')
ax.set_xlim(0, correlation_plot_rmax)
ax.legend(fontsize=8)
ax.grid(alpha=0.25)
plt.tight_layout()
fig.savefig(
    validation_figure_path('scalar_xi3'), dpi=160,
    bbox_inches='tight'
)
plt.show()

sector_component_list = [
    xi3_KTZ_bin_components(
        r_lower, r_upper, phase_limited=True
    )
    for r_lower, r_upper in zip(sector_r_lower, sector_r_upper)
]
sector_ktz_components = components_to_arrays(sector_component_list)
sector_ktz = sum(
    sector_ktz_components[name] for name in component_order
)
sector_CPN_max = np.nanmax(np.abs(sector_ktz_components['CPN']))
assert sector_CPN_max > 1.0e-3
sector_comp0 = xi3_KTZ_components(0.0, phase_limited=True)
sector_comp_plot = {
    name: np.concatenate([[sector_comp0[name]], sector_ktz_components[name]])
    for name in component_order
}
sector_ktz_plot = sum(
    sector_comp_plot[name] for name in component_order
)
sector_r_plot = np.concatenate([[0.0], sector_r_centers])
sector_sim_plot = np.concatenate([
    [1.0], sector_sim_mean[sector_bin_mask]
])
sector_std_plot = np.concatenate([
    [0.0], sector_sim_std[sector_bin_mask]
])
sector_sem_plot = np.concatenate([
    [0.0], sector_sim_sem[sector_bin_mask]
])

fig, ax = plt.subplots(figsize=(7, 5))
plot_theory_component_stack(ax, sector_r_plot, sector_comp_plot)
ax.plot(
    sector_r_plot, sector_sim_plot, 'o-', color='black', ms=4,
    label='Simulation mean', zorder=4
)
ax.fill_between(
    sector_r_plot, sector_sim_plot-sector_std_plot,
    sector_sim_plot+sector_std_plot, color='black', alpha=0.14,
    linewidth=0, label=r'Single-realisation $1\sigma$', zorder=3
)
ax.errorbar(
    sector_r_plot, sector_sim_plot, yerr=sector_sem_plot, fmt='none',
    ecolor='black', elinewidth=0.9, capsize=2,
    label=r'Mean $1\sigma$ uncertainty', zorder=5
)
ax.plot(
    sector_r_plot, sector_ktz_plot, 'k--', lw=1.5,
    label='KTZ total', zorder=6
)
ax.axhline(0, color='grey', lw=0.5)
ax.set_xlabel('r (kpc)')
ax.set_ylabel(r'$\xi_{\rm arm,pp}(r)$')
ax.set_xlim(0, sector_validation_rmax)
ax.set_title(
    r'Arm-phase sector: $|\Theta| < 30^\circ$, '
    rf'${sector_analysis_R_in:.1f}<R_m<{sector_analysis_R_out:.1f}$ kpc'
)
ax.legend(fontsize=8)
ax.grid(alpha=0.25)
plt.tight_layout()
fig.savefig(
    validation_figure_path('scalar_sector'), dpi=160,
    bbox_inches='tight'
)
plt.show()

xi3_validation_metrics = summarize_curve_validation(
    xi3_sim_mean[validation_bin_mask],
    xi3_ktz,
    xi3_sim_sem[validation_bin_mask],
)
sector_validation_metrics = summarize_curve_validation(
    sector_sim_mean[sector_bin_mask],
    sector_ktz,
    sector_sim_sem[sector_bin_mask],
)
validation_summaries = [('xi3', xi3_validation_metrics)]
if not enable_source_clustering:
    validation_summaries.append(('arm-sector', sector_validation_metrics))
for label, metrics in validation_summaries:
    print(
        f'{label} validation (N={metrics["n"]}): '
        f'RMSE={metrics["rmse"]:.3f}, '
        f'mean(simulation - KTZ)={metrics["bias"]:+.3f}, '
        f'Pearson r={metrics["pearson"]:.3f}, '
        f'max|residual|={metrics["max_abs"]:.3f}, '
        f'within mean 1 sigma={metrics["within_1sigma"]}/'
        f'{metrics["n"]}'
    )
if enable_source_clustering:
    print('arm-sector result excluded from the isotropic CNP validation')
else:
    print(f'arm-sector max|CPN contribution|={sector_CPN_max:.3f}')
```

```python
# %% Cross-correlation Block 1: Load the KT25 yield database
from pathlib import Path
import os
import sys
import astropy.units as u
from scipy.integrate import simpson
from matplotlib.colors import TwoSlopeNorm

NOTEBOOK_ROOT = Path.cwd()

LOCAL_KT25_REPO = NOTEBOOK_ROOT / 'data' / 'kt25'
default_kt25_repo = (
    LOCAL_KT25_REPO
    if (LOCAL_KT25_REPO / 'process_data.py').exists()
    else Path.home() / '.local' / 'share' / 'z_fluctuation'
)
KT25_REPO = Path(os.environ.get(
    'KT25_REPO', str(default_kt25_repo)
))

SLUG2_REPO = Path(os.environ.get(
    'SLUG2_REPO', Path.home() / '.local' / 'share' / 'slug2'
))
SLUG_STOCHASTIC_DIR = Path(os.environ.get(
    'SLUG_STOCHASTIC_DIR', KT25_REPO / 'slug_stochastic'
))
SLUG_MOMENT_FILE = SLUG_STOCHASTIC_DIR / 'kt25_slug_yield_moments.npz'

if not (KT25_REPO / 'process_data.py').exists():
    raise FileNotFoundError(f'KT25 repository not found at {KT25_REPO}')
if not (SLUG2_REPO / 'slugpy').exists():
    raise FileNotFoundError(f'slugpy not found at {SLUG2_REPO}')

for module_path in (SLUG2_REPO, KT25_REPO):
    module_path = str(module_path)
    if module_path not in sys.path:
        sys.path.insert(0, module_path)

from slugpy import read_cluster
from asplund09_abd import solar_abd
from seitenzahl13_yld import s13_yld

kt25_channels = ['snii', 'agb', 'snia']
kt25_r_process = [
    'Ru', 'Rh', 'Rd', 'Ag', 'Cd', 'In', 'Sb', 'Te', 'I', 'Xe',
    'Cs', 'La', 'Pr', 'Nd', 'Sm', 'Eu', 'Gd', 'Tb', 'Dy',
    'Ho', 'Er', 'Tm', 'Yb'
]

# Yield and delay-time ingestion following the KT25 reduction, so that the
# channel yields and t_50 delays match KT25.
kt25_prod = {
    row['sym']: {}
    for row in solar_abd
}
slug_agb = read_cluster(str(KT25_REPO / 'slug' / 'yields_agb'))
slug_snii = read_cluster(str(KT25_REPO / 'slug' / 'yields_snii'))
kt25_time = slug_agb.time * u.yr

kt25_tmin = 40 * u.Myr
kt25_tmax = 13.7 * u.Gyr
kt25_dtd_alpha = -1.1
kt25_dtd_norm = (
    1.3e-3 * (1 + kt25_dtd_alpha)
    / (
        kt25_tmax * (kt25_tmax / kt25_tmin)**kt25_dtd_alpha
        - kt25_tmin
    )
)
kt25_nIa = kt25_dtd_norm * (
    kt25_time * (kt25_time / kt25_tmin)**kt25_dtd_alpha - kt25_tmin
) / (1 + kt25_dtd_alpha)
kt25_nIa[kt25_time < kt25_tmin] = 0

for element in kt25_prod:
    channel_yields = {
        name: np.zeros(kt25_time.size)
        for name in ['tot', 'snii', 'agb', 'snia']
    }

    for idx in np.where(slug_snii.isotope_name == element)[0]:
        channel_yields['snii'] += slug_snii.yld[:, idx]
        channel_yields['tot'] += slug_snii.yld[:, idx]

    for idx in np.where(slug_agb.isotope_name == element)[0]:
        channel_yields['agb'] += slug_agb.yld[:, idx]
        channel_yields['tot'] += slug_agb.yld[:, idx]

    for idx in np.where(np.asarray(s13_yld['sym']) == element)[0]:
        ia_yield = s13_yld['ylds']['N100'][idx] * kt25_nIa
        channel_yields['snia'] += ia_yield
        channel_yields['tot'] += ia_yield

    for name, values in channel_yields.items():
        kt25_prod[element]['yld_' + name] = values

for element in kt25_prod:
    if kt25_prod[element]['yld_tot'][-1] <= 0:
        continue

    for channel in kt25_channels:
        channel_yield = kt25_prod[element]['yld_' + channel]
        kt25_prod[element]['t50_' + channel] = float(
            kt25_time[np.argmax(
                channel_yield > 0.5 * channel_yield[-1]
            )].to(u.Gyr).value
        )

kt25_elements = [
    row['sym'] for row in solar_abd
    if 6 <= row['Z'] <= 60
    and row['sym'] not in kt25_r_process
    and kt25_prod[row['sym']]['yld_tot'][-1] > 0
]
# Mean returned mass of each element and channel per unit stellar mass formed.
# The SLUG models use cluster_mass=1, so the final cumulative yields are
# directly the first yield moments needed by the covariance calculation.
kt25_mean_yield = np.array([
    [kt25_prod[element]['yld_' + channel][-1] for channel in kt25_channels]
    for element in kt25_elements
], dtype=float)
kt25_delay = np.array([
    [kt25_prod[element]['t50_' + channel] for channel in kt25_channels]
    for element in kt25_elements
])

if not SLUG_MOMENT_FILE.exists():
    raise FileNotFoundError(
        f'Stochastic SLUG moments not found at {SLUG_MOMENT_FILE}. '
        'Run slug_stochastic/build_yield_moments.py first.'
    )

slug_moments = np.load(SLUG_MOMENT_FILE, allow_pickle=False)
if list(slug_moments['elements']) != kt25_elements:
    raise ValueError('Stochastic SLUG element ordering does not match KT25')
if list(slug_moments['channels']) != kt25_channels:
    raise ValueError('Stochastic SLUG channel ordering does not match KT25')

kt25_imf_cov_per_mass = slug_moments['imf_covariance_per_mass']
kt25_event_mass_mean = float(slug_moments['event_mass_mean'])
kt25_event_mass_second = float(slug_moments['event_mass_second'])
kt25_sigma_w2 = float(slug_moments['sigma_w2'])
# The analytic prediction uses the same-event second moment, not the
# variance alone. Define the event-specific yield fraction by
# M_* y_Xa^(e) = M_* <y_Xa> + epsilon_Xa, with
# E[epsilon | M_*] = 0 and Cov(epsilon | M_*) = M_* K. Then
#
# E[M_*^2 y_Xa^(e) y_Yb^(e)]
#     = E[M_*^2] <y_Xa><y_Yb> + E[M_*] K_Xa,Yb.
#
# Dividing by E[M_*^2] gives the channel-pair weight that replaces the
# deterministic product <y_Xa><y_Yb> in C_2p,0 and C_2p,N.
kt25_mean_yield_product = np.einsum(
    'ia,jb->iajb', kt25_mean_yield, kt25_mean_yield
)
kt25_event_yield_second_moment = (
    kt25_event_mass_second * kt25_mean_yield_product
    + kt25_event_mass_mean * kt25_imf_cov_per_mass
)
kt25_second_yield_normalized = (
    kt25_event_yield_second_moment / kt25_event_mass_second
)

# Fiducial KT25 parameters.
_km_s_to_kpc_gyr = (1 * u.km / u.s).to(u.kpc / u.Gyr).value
kt25_lcorr = 1.0
kt25_kappa = (150 * u.pc * 7 * u.km / u.s / 3).to(u.kpc**2 / u.Gyr).value
kt25_sigma_v = 2.0 * _km_s_to_kpc_gyr
kt25_t_corr = kt25_lcorr**2 / kt25_kappa
kt25_Gamma = (3e-6 / u.Myr / u.pc**2).to(1 / u.Gyr / u.kpc**2).value
kt25_sigma_inj = np.array([0.060, 0.001, 0.060])
```

### KT25 and stochastic-SLUG inputs

The calculation uses the KT25 yield database and paired stochastic SLUG2 realizations. Its principal parameters are
$$
t_{\rm corr}=2.794\ {\rm Gyr},\qquad
\kappa=0.3579\ {\rm kpc^2\,Gyr^{-1}},\qquad
\sigma_v=2.0454\ {\rm kpc\,Gyr^{-1}},
$$
$$
(\sigma_{\rm CCSN},\sigma_{\rm AGB},\sigma_{\rm TNSN})
=(60,1,60)\ {\rm pc}.
$$

Here $t_{\rm corr}\equiv\ell_{\rm corr}^2/\kappa$ follows the KT25 length convention. It is not the single-element $\ell_\ast$: if $t_\ast=t_{\rm corr}$ and a common injection width $\sigma$ were used, then $\ell_\ast=(\sigma^2+2\ell_{\rm corr}^2)^{1/2}$.

For a cluster mass distribution $\mathrm{d}\mathcal N/\mathrm{d}M_\star\propto M_\star^{-2}$ over $10^2\leq M_\star/M_\odot\leq10^6$, the same-event channel weight is
$$
\frac{\langle M_\star^2 y_{A,a}^{(e)}y_{B,b}^{(e)}\rangle}{\langle M_\star^2\rangle}
=y_{A,a}y_{B,b}
+\frac{\langle M_\star\rangle}{\langle M_\star^2\rangle}K_{Aa,Bb}.
$$


```python
# %% Cross-correlation Block 2: Reproduce the KT25 baseline
def _kt25_kernel(alpha, beta, gamma):
    return (
        np.log(alpha)
        + 2 * beta / np.sqrt(1 + beta**2) * np.arcsinh(beta)
        - 2 * gamma / np.sqrt(1 + gamma**2) * np.arcsinh(gamma)
    )


def kt25_covariance_matrix(
    sigma_inj, channel_yields=None, channel_second_moment=None
):
    if channel_second_moment is None:
        if channel_yields is None:
            raise ValueError(
                'Provide channel_yields or channel_second_moment'
            )
        channel_second_moment = np.einsum(
            'ia,jb->iajb', channel_yields, channel_yields
        )
    elif channel_yields is not None:
        raise ValueError(
            'channel_yields and channel_second_moment are mutually exclusive'
        )

    ne = len(kt25_elements)
    covariance = np.zeros((ne, ne))

    for i in range(ne):
        for j in range(i, ne):
            value = 0.0
            for a in range(len(kt25_channels)):
                delay_a = kt25_delay[i, a]
                ell_a2 = kt25_kappa * delay_a

                for b in range(len(kt25_channels)):
                    yield_second = channel_second_moment[i, a, j, b]
                    if yield_second == 0:
                        continue
                    delay_b = kt25_delay[j, b]
                    ell_b2 = kt25_kappa * delay_b
                    sigma2 = sigma_inj[a]**2 + sigma_inj[b]**2
                    delay_difference = abs(delay_a - delay_b)

                    numerator = (
                        4 * kt25_lcorr**2 - 2 * ell_a2 - 2 * ell_b2
                        + sigma2
                    )
                    denominator = 2 * abs(ell_a2 - ell_b2) + sigma2
                    alpha = numerator / denominator
                    beta = (
                        kt25_sigma_v * delay_difference / np.sqrt(numerator)
                    )
                    gamma = (
                        kt25_sigma_v * delay_difference / np.sqrt(denominator)
                    )
                    value += yield_second * _kt25_kernel(
                        alpha, beta, gamma
                    )

            covariance[i, j] = value
            covariance[j, i] = value

    return covariance


def normalize_cross_covariance(covariance):
    variance = np.diag(covariance)
    denominator = np.sqrt(np.outer(variance, variance))
    return np.divide(
        covariance, denominator,
        out=np.full_like(covariance, np.nan),
        where=denominator > 0
    )


# Spatially uniform KT25 baseline with the published channel widths.
kt25_CP0 = kt25_covariance_matrix(
    kt25_sigma_inj, channel_yields=kt25_mean_yield
)
Xi_KT25 = normalize_cross_covariance(kt25_CP0)

assert np.nanmax(np.abs(Xi_KT25 - Xi_KT25.T)) < 1.0e-12
assert np.nanmax(np.abs(np.diag(Xi_KT25) - 1.0)) < 1.0e-12
```

```python
# %% Cross-correlation Block 3: Full-window non-uniform KTZ prediction
cross_dx = 0.05  # kpc
cross_window_margin = np.sqrt(2.0) * cross_dx
cross_analysis_R_in = R_in + cross_window_margin
cross_analysis_R_out = R_out - cross_window_margin
cross_radial_nodes = 30
cross_spectrum_nodes = 1000

cross_rho2_edges = np.linspace(
    cross_analysis_R_in**2,
    cross_analysis_R_out**2,
    cross_radial_nodes + 1
)
cross_varpi_nodes = np.sqrt(
    0.5 * (cross_rho2_edges[:-1] + cross_rho2_edges[1:])
)
cross_rate_relative = np.exp(-cross_varpi_nodes / h_R)
cross_AP_weights = cross_rate_relative.copy()
cross_AP_weight_sum = np.sum(cross_AP_weights)
cross_prefactor_0 = (
    8 * np.pi * kt25_kappa * kt25_Gamma
    / (1 + kt25_sigma_w2)
)


def _cross_harmonic_damping(Qn):
    if not enable_source_clustering or cluster_fraction <= 0:
        return 1.0
    return (
        1.0 - cluster_fraction
        + cluster_fraction
        * np.exp(-0.5 * (Qn * cluster_radius)**2)
    )


def _mean_transfer(delay, sigma, Q):
    if delay >= kt25_t_corr:
        return 0.0
    duration = kt25_t_corr - delay
    diffusion = -np.expm1(
        -kt25_kappa * duration * Q**2
    ) / (kt25_kappa * Q**2)
    injection = np.exp(-0.5 * (sigma * Q)**2)
    drift = np.exp(
        -0.5 * (kt25_sigma_v * delay * Q)**2
    )
    return injection * drift * diffusion


def _square_pixel_overlap_1d(combined_sigma, pixel_size):
    """Zero-lag overlap of two square-pixel-averaged 1D kernels."""
    ratio = pixel_size / combined_sigma
    first = pixel_size * (ndtr(ratio) - 0.5)
    second = (
        combined_sigma / np.sqrt(2 * np.pi)
        * (-np.expm1(-0.5 * ratio**2))
    )
    return 2.0 * (first - second) / pixel_size**2


def kt25_square_pixel_covariance_matrix(
    pixel_size, channel_second_moment
):
    """Same-parent covariance with the exact square-pixel response."""
    covariance = np.zeros((len(kt25_elements), len(kt25_elements)))
    kernel_cache = {}

    def channel_kernel(delay_a, delay_b, sigma_a, sigma_b):
        key = (delay_a, delay_b, sigma_a, sigma_b)
        if key in kernel_cache:
            return kernel_cache[key]
        lower = max(delay_a, delay_b)
        if lower >= kt25_t_corr:
            value = 0.0
        else:
            drift_variance = (
                kt25_sigma_v * (delay_a - delay_b)
            )**2

            def age_integrand(parent_age):
                combined_variance = (
                    sigma_a**2 + sigma_b**2
                    + 2 * kt25_kappa
                    * (2 * parent_age - delay_a - delay_b)
                    + drift_variance
                )
                overlap_1d = _square_pixel_overlap_1d(
                    np.sqrt(combined_variance), pixel_size
                )
                return overlap_1d**2

            age_integral, _ = quad(
                age_integrand, lower, kt25_t_corr,
                epsabs=1.0e-10, epsrel=2.0e-8, limit=200
            )
            value = 8 * np.pi * kt25_kappa * age_integral
        kernel_cache[key] = value
        kernel_cache[(delay_b, delay_a, sigma_b, sigma_a)] = value
        return value

    for i in range(len(kt25_elements)):
        for j in range(i, len(kt25_elements)):
            value = 0.0
            for a in range(len(kt25_channels)):
                for b in range(len(kt25_channels)):
                    yield_second = channel_second_moment[i, a, j, b]
                    if yield_second == 0:
                        continue
                    value += yield_second * channel_kernel(
                        kt25_delay[i, a], kt25_delay[j, b],
                        kt25_sigma_inj[a], kt25_sigma_inj[b]
                    )
            covariance[i, j] = value
            covariance[j, i] = value
    return covariance


def _square_pixel_power_average(Q, pixel_size, n_angle=64):
    """Orientation-averaged squared square-pixel Fourier window."""
    if pixel_size is None:
        return np.ones_like(Q, dtype=float)
    Q_array = np.asarray(Q, dtype=float)
    angle = np.linspace(0.0, 2 * np.pi, n_angle, endpoint=False)
    kx = Q_array[..., None] * np.cos(angle)
    ky = Q_array[..., None] * np.sin(angle)
    window = (
        np.sinc(kx * pixel_size / (2 * np.pi))
        * np.sinc(ky * pixel_size / (2 * np.pi))
    )
    return np.mean(window**2, axis=-1)


ne = len(kt25_elements)
k_cross = np.geomspace(1.0e-4, 80.0, cross_spectrum_nodes)
if (
    enable_source_clustering
    and cluster_fraction > 0
    and cluster_radius > 0
):
    P_source = (
        cluster_fraction * mean_children_per_cluster
        / (kt25_Gamma * kt25_t_corr)
        * np.exp(-(cluster_radius * k_cross)**2)
    )
else:
    P_source = np.zeros_like(k_cross)


def _ktz_cnp_matrix(sigma_inj, pixel_size=None):
    transfer = np.zeros((ne, len(k_cross)))
    for element_index in range(ne):
        for channel_index in range(len(kt25_channels)):
            yield_value = kt25_mean_yield[
                element_index, channel_index
            ]
            delay = kt25_delay[element_index, channel_index]
            if yield_value == 0 or delay >= kt25_t_corr:
                continue
            duration = kt25_t_corr - delay
            diffusion = -np.expm1(
                -kt25_kappa * duration * k_cross**2
            ) / (kt25_kappa * k_cross**2)
            injection = np.exp(
                -0.5 * (sigma_inj[channel_index] * k_cross)**2
            )
            drift = np.exp(
                -0.5 * (kt25_sigma_v * delay * k_cross)**2
            )
            transfer[element_index] += (
                yield_value * injection * drift * diffusion
            )

    covariance = np.zeros((ne, ne))
    spectral_weight = (
        k_cross * P_source / (2 * np.pi)
        * _square_pixel_power_average(k_cross, pixel_size)
    )
    for i in range(ne):
        for j in range(i, ne):
            value = cross_prefactor_0 * simpson(
                spectral_weight * transfer[i] * transfer[j],
                x=k_cross
            )
            covariance[i, j] = value
            covariance[j, i] = value
    return covariance


def compute_ktz_window_components(
    sigma_inj, pixel_size=None, cp0_override=None
):
    cp0 = (
        kt25_covariance_matrix(
            sigma_inj,
            channel_second_moment=kt25_second_yield_normalized
        )
        if cp0_override is None else cp0_override
    )
    cnp = _ktz_cnp_matrix(sigma_inj, pixel_size=pixel_size)
    components = {
        'CP0': np.zeros_like(cp0),
        'CPN': np.zeros_like(cp0),
        'CNP': np.zeros_like(cp0),
        'CM': np.zeros_like(cp0),
    }

    for varpi, rate_relative, AP_weight in zip(
        cross_varpi_nodes,
        cross_rate_relative,
        cross_AP_weights
    ):
        q_base = k_perp(varpi)
        local_cm = np.zeros_like(cp0)
        local_prefactor = (
            cross_prefactor_0 * rate_relative
        )
        for n, g_n, alpha_n in zip(
            harmonic_n, harmonic_g, harmonic_alpha
        ):
            q_n = n * q_base
            amplitude = (
                eta * g_n * _cross_harmonic_damping(q_n)
            )
            mean_vector = np.zeros(ne)
            for element_index in range(ne):
                for channel_index in range(
                    len(kt25_channels)
                ):
                    yield_value = kt25_mean_yield[
                        element_index, channel_index
                    ]
                    if yield_value == 0:
                        continue
                    mean_vector[element_index] += (
                        yield_value * _mean_transfer(
                            kt25_delay[
                                element_index, channel_index
                            ],
                            sigma_inj[channel_index],
                            q_n
                        )
                    )
            modulated_mean = amplitude * mean_vector
            pixel_power = _square_pixel_power_average(
                q_n, pixel_size
            )
            local_cm += (
                0.5 * local_prefactor
                * pixel_power
                * np.outer(modulated_mean, modulated_mean)
            )

        components['CP0'] += AP_weight * cp0
        components['CNP'] += AP_weight * cnp
        components['CM'] += AP_weight * local_cm

    for name in components:
        components[name] /= cross_AP_weight_sum
    return components


ktz_window_components = compute_ktz_window_components(
    kt25_sigma_inj
)
Cov_KTZ_window = sum(ktz_window_components.values())
Xi_KTZ_window = normalize_cross_covariance(
    Cov_KTZ_window
)

assert np.count_nonzero(
    ktz_window_components['CPN']
) == 0
assert np.nanmax(
    np.abs(Xi_KTZ_window - Xi_KTZ_window.T)
) < 1.0e-12
assert np.nanmax(
    np.abs(np.diag(Xi_KTZ_window) - 1.0)
) < 1.0e-12
```

## Numerical validation of elemental cross-correlation

The simulation samples the continuous KT25 parent-event process and deposits the returned fields into $50\,\mathrm{pc}$ square pixels. The stochastic yield-residual vector is drawn from a multivariate Gaussian that matches the first two moments estimated from paired stochastic SLUG2 realizations. All elements share parent positions, event masses, stellar velocities, and the Thomas-process parent complexes; channel delays, injection widths, drift, diffusion, and stochastic yields determine their return fields. Event kernels narrower than three pixels retain their individual ages and continuous return positions and are integrated over the square pixel faces with the Gaussian CDF. Broader event kernels are grouped into 40 bins uniform in the logarithm of squared width and evaluated by convolution on the pixel-centre mesh. This narrow--broad division is a numerical acceleration rather than a physical distinction. The narrow branch is analytic within the adopted five-dispersion support; the broad branch is a controlled approximation to the same pixel-averaged field. The analytic $C_{{\rm 2p},0}$, $C_{\rm NP}$, and mean-field terms use the corresponding exact square-pixel overlap or orientation-averaged square window; no Gaussian variance approximation is used.
The reported simulation coefficient is formed by normalising the covariance averaged over the 12 realisations. Error bars are bootstrap standard errors from 1000 resamples of those covariance matrices. Because the simulation--KTZ comparison is already shown by the scatter plot, the matrix comparison below instead shows the spatially uniform KT25 baseline, the pixel-matched KTZ prediction, and their KTZ-minus-KT25 difference.


```python
# %% Cross-correlation Block 6: Multi-element KTZ numerical realizations
from scipy.ndimage import gaussian_filter

cross_validation_elements = cross_plot_elements.copy()
cross_validation_index = np.array([
    kt25_elements.index(element)
    for element in cross_validation_elements
])
cross_saved_elements = ['O', 'N']
cross_saved_index = np.array([
    cross_validation_elements.index(element)
    for element in cross_saved_elements
])
cross_validation_n_realizations = 12
cross_age_bins = 40
# Kernels narrower than three pixels are integrated over the square pixel
# faces at their continuous return positions and with their individual ages.
# Broader kernels retain logarithmic width bins and fast Gaussian filtering;
# at sigma >= 3*dx the nearest-cell approximation is below the percent level
# in the controlled deposition test. The analytic prediction uses the exact
# square-pixel overlap rather than a Gaussian variance approximation.
cross_exact_sigma_threshold = 3.0 * cross_dx
cross_source_padding = 6.0
cross_source_R_in = max(0.30, R_in - cross_source_padding)
cross_source_R_out = R_out + cross_source_padding
cross_grid_radius = cross_source_R_out
cross_grid_x = np.arange(
    -cross_grid_radius,
    cross_grid_radius + cross_dx,
    cross_dx
)
cross_grid_y = cross_grid_x.copy()
cross_X, cross_Y = np.meshgrid(cross_grid_x, cross_grid_y)
cross_R = np.hypot(cross_X, cross_Y)
cross_disk_mask = (cross_R >= R_in) & (cross_R <= R_out)
cross_analysis_mask = (
    (cross_R >= cross_analysis_R_in)
    & (cross_R <= cross_analysis_R_out)
)

def _radial_rate_integral(radius_min, radius_max):
    def primitive(radius):
        return -h_R * (radius + h_R) * np.exp(-radius / h_R)
    return primitive(radius_max) - primitive(radius_min)


cross_expected_parent_events = (
    2 * np.pi * kt25_Gamma * kt25_t_corr
    * _radial_rate_integral(cross_source_R_in, cross_source_R_out)
)
# Inverse CDF for the axisymmetric radial density
# p(varpi) proportional to varpi exp(-varpi/h_R).
_cross_radius_table = np.linspace(
    cross_source_R_in, cross_source_R_out, 20000
)
_cross_radius_cdf = np.array([
    _radial_rate_integral(cross_source_R_in, radius)
    for radius in _cross_radius_table
])
_cross_radius_cdf /= _cross_radius_cdf[-1]
_cross_rate_contrast_max = np.max(
    1.0 + eta * h_func(Theta_check)
)


def sample_cross_parent_positions(n_sample, rng):
    """Sample the KTZ spiral-modulated parent-event intensity."""
    if n_sample <= 0:
        return np.empty(0), np.empty(0)

    radius = np.empty(0)
    phi = np.empty(0)
    while len(radius) < n_sample:
        n_needed = n_sample - len(radius)
        n_candidate = max(2 * n_needed, 256)
        candidate_radius = np.interp(
            rng.uniform(size=n_candidate),
            _cross_radius_cdf,
            _cross_radius_table
        )
        candidate_phi = rng.uniform(
            0.0, 2 * np.pi, n_candidate
        )
        theta = spiral_phase(candidate_radius, candidate_phi)
        contrast = 1.0 + eta * h_func(theta)
        accept = (
            rng.uniform(size=n_candidate)
            < contrast / _cross_rate_contrast_max
        )
        radius = np.concatenate((radius, candidate_radius[accept]))
        phi = np.concatenate((phi, candidate_phi[accept]))

    return radius[:n_sample], phi[:n_sample]


def sample_cross_clustered_positions(rng):
    """Draw an exact Thomas component with a compound-Poisson child count."""
    expected_parent_complexes = (
        cluster_fraction * cross_expected_parent_events
        / mean_children_per_cluster
    )
    n_parent = rng.poisson(expected_parent_complexes)
    if n_parent == 0:
        return np.empty(0), np.empty(0)

    parent_radius, parent_phi = sample_cross_parent_positions(
        n_parent, rng
    )
    parent_x = parent_radius * np.cos(parent_phi)
    parent_y = parent_radius * np.sin(parent_phi)
    child_count = rng.poisson(
        mean_children_per_cluster, n_parent
    )
    n_child = int(np.sum(child_count))
    if n_child == 0:
        return np.empty(0), np.empty(0)

    child_x = (
        np.repeat(parent_x, child_count)
        + rng.normal(0.0, cluster_radius, n_child)
    )
    child_y = (
        np.repeat(parent_y, child_count)
        + rng.normal(0.0, cluster_radius, n_child)
    )
    child_radius = np.hypot(child_x, child_y)
    retained = (
        (child_radius >= cross_source_R_in)
        & (child_radius <= cross_source_R_out)
    )
    return child_x[retained], child_y[retained]


def sample_event_mass(n_event, rng):
    """Sample p(M_star) proportional to M_star^-2."""
    inverse_mass = (
        1.0 / float(slug_moments['cmf_min'])
        - rng.uniform(size=n_event)
        * (
            1.0 / float(slug_moments['cmf_min'])
            - 1.0 / float(slug_moments['cmf_max'])
        )
    )
    return 1.0 / inverse_mass


cross_mean_yield = kt25_mean_yield[cross_validation_index]
cross_delay = kt25_delay[cross_validation_index]
cross_flat_index = np.array([
    element_index * len(kt25_channels) + channel_index
    for element_index in cross_validation_index
    for channel_index in range(len(kt25_channels))
])
cross_K = kt25_imf_cov_per_mass.reshape(
    len(kt25_elements) * len(kt25_channels),
    len(kt25_elements) * len(kt25_channels)
)[np.ix_(cross_flat_index, cross_flat_index)]
cross_K = 0.5 * (cross_K + cross_K.T)
cross_K_eigenvalue, cross_K_eigenvector = np.linalg.eigh(cross_K)
cross_K_factor = (
    cross_K_eigenvector
    * np.sqrt(np.clip(cross_K_eigenvalue, 0.0, None))
)


def sample_cross_event_yields(event_mass, rng):
    """Sample joint channel yields with the SLUG same-event second moment."""
    n_event = len(event_mass)
    mean_flat = cross_mean_yield.reshape(-1)
    stochastic_residual = (
        rng.normal(size=(n_event, len(mean_flat)))
        @ cross_K_factor.T
    )
    event_yield = (
        event_mass[:, None] * mean_flat[None, :]
        + np.sqrt(event_mass)[:, None] * stochastic_residual
    )
    return event_yield.reshape(
        n_event,
        len(cross_validation_elements),
        len(kt25_channels)
    )


def _deposit_binned_gaussian(return_x, return_y, weight, sigma):
    """Fast broad-kernel deposition for one logarithmic width bin."""
    nx = len(cross_grid_x)
    ny = len(cross_grid_y)
    ix = np.floor(
        (return_x - cross_grid_x[0]) / cross_dx + 0.5
    ).astype(int)
    iy = np.floor(
        (return_y - cross_grid_y[0]) / cross_dx + 0.5
    ).astype(int)
    inside = (
        (ix >= 0) & (ix < nx)
        & (iy >= 0) & (iy < ny)
        & np.isfinite(weight)
    )
    source = np.bincount(
        iy[inside] * nx + ix[inside],
        weights=weight[inside],
        minlength=nx * ny
    ).reshape(ny, nx)
    return gaussian_filter(
        source,
        sigma=max(sigma / cross_dx, 1.0e-3),
        mode='constant',
        cval=0.0,
        truncate=5.0
    ) / cross_dx**2


def _deposit_pixel_integrated(
    return_x, return_y, weight, sigma, cutoff=5.0
):
    """Integrate continuous Gaussian kernels over square pixel faces."""
    result = np.zeros_like(cross_X, dtype=float)
    half_pixel = 0.5 * cross_dx
    nx = len(cross_grid_x)
    ny = len(cross_grid_y)
    sigma = np.broadcast_to(
        np.asarray(sigma, dtype=float), np.shape(return_x)
    )

    for event_x, event_y, event_weight, event_sigma in zip(
        return_x, return_y, weight, sigma
    ):
        if (
            not np.isfinite(event_weight)
            or not np.isfinite(event_sigma)
            or event_sigma <= 0
        ):
            continue
        support = cutoff * event_sigma + half_pixel
        ix0 = max(0, int(np.ceil(
            (event_x - support - cross_grid_x[0]) / cross_dx
        )))
        ix1 = min(nx - 1, int(np.floor(
            (event_x + support - cross_grid_x[0]) / cross_dx
        )))
        iy0 = max(0, int(np.ceil(
            (event_y - support - cross_grid_y[0]) / cross_dx
        )))
        iy1 = min(ny - 1, int(np.floor(
            (event_y + support - cross_grid_y[0]) / cross_dx
        )))
        if ix0 > ix1 or iy0 > iy1:
            continue

        pixel_x = cross_grid_x[ix0:ix1 + 1]
        pixel_y = cross_grid_y[iy0:iy1 + 1]
        probability_x = (
            ndtr((pixel_x + half_pixel - event_x) / event_sigma)
            - ndtr((pixel_x - half_pixel - event_x) / event_sigma)
        )
        probability_y = (
            ndtr((pixel_y + half_pixel - event_y) / event_sigma)
            - ndtr((pixel_y - half_pixel - event_y) / event_sigma)
        )
        result[iy0:iy1 + 1, ix0:ix1 + 1] += (
            event_weight
            * np.outer(probability_y, probability_x)
            / cross_dx**2
        )

    return result


def cross_axisymmetric_mean_fields():
    """Return the axisymmetric ensemble means on the cross pixel-centre mesh."""
    safe_radius = np.maximum(cross_R, cross_source_R_in)
    local_rate = kt25_Gamma * np.exp(-safe_radius / h_R)
    axisymmetric = np.zeros(
        (len(cross_validation_elements), *cross_R.shape)
    )
    for element_index in range(len(cross_validation_elements)):
        response = 0.0
        for channel_index in range(len(kt25_channels)):
            delay = cross_delay[element_index, channel_index]
            if delay < kt25_t_corr:
                response += (
                    cross_mean_yield[element_index, channel_index]
                    * (kt25_t_corr - delay)
                )
        axisymmetric[element_index] = (
            local_rate * kt25_event_mass_mean * response
        )
    return axisymmetric


cross_axisymmetric_mean = cross_axisymmetric_mean_fields()


def generate_cross_ktz_realization(rng):
    """Generate one multi-element KTZ realization and its residual fields."""
    if enable_source_clustering and cluster_fraction > 0:
        f_cl = np.clip(cluster_fraction, 0.0, 1.0)
        n_poisson = rng.poisson(
            (1.0 - f_cl) * cross_expected_parent_events
        )
        clustered_x, clustered_y = sample_cross_clustered_positions(rng)
    else:
        n_poisson = rng.poisson(cross_expected_parent_events)
        clustered_x = np.empty(0)
        clustered_y = np.empty(0)

    poisson_radius, poisson_phi = sample_cross_parent_positions(
        n_poisson, rng
    )
    poisson_x = poisson_radius * np.cos(poisson_phi)
    poisson_y = poisson_radius * np.sin(poisson_phi)
    event_x = np.concatenate((poisson_x, clustered_x))
    event_y = np.concatenate((poisson_y, clustered_y))
    n_event = len(event_x)
    permutation = rng.permutation(n_event)
    event_x = event_x[permutation]
    event_y = event_y[permutation]

    event_age = rng.uniform(0.0, kt25_t_corr, n_event)
    velocity_x = rng.normal(0.0, kt25_sigma_v, n_event)
    velocity_y = rng.normal(0.0, kt25_sigma_v, n_event)
    event_mass = sample_event_mass(n_event, rng)
    event_yield = sample_cross_event_yields(event_mass, rng)

    fields = np.zeros(
        (len(cross_validation_elements), *cross_R.shape)
    )
    for element_index in range(len(cross_validation_elements)):
        for channel_index in range(len(kt25_channels)):
            delay = cross_delay[element_index, channel_index]
            duration = kt25_t_corr - delay
            if (
                duration <= 0
                or not np.any(
                    event_yield[:, element_index, channel_index] != 0
                )
            ):
                continue

            return_x = event_x + velocity_x * delay
            return_y = event_y + velocity_y * delay
            time_since_return = event_age - delay
            sigma_injection2 = kt25_sigma_inj[channel_index]**2
            valid_return = (
                (time_since_return >= 0.0)
                & (time_since_return < duration)
            )
            event_width2 = (
                sigma_injection2
                + 2 * kt25_kappa * np.maximum(time_since_return, 0.0)
            )
            exact_selected = (
                valid_return
                & (
                    event_width2
                    < cross_exact_sigma_threshold**2
                )
            )
            if np.any(exact_selected):
                fields[element_index] += _deposit_pixel_integrated(
                    return_x[exact_selected],
                    return_y[exact_selected],
                    event_yield[
                        exact_selected, element_index, channel_index
                    ],
                    np.sqrt(event_width2[exact_selected])
                )

            # Broad kernels are smooth on the pixel-centre mesh. Retain logarithmic
            # width bins for these events and use the geometric midpoint.
            width2_edges = np.geomspace(
                sigma_injection2,
                sigma_injection2 + 2 * kt25_kappa * duration,
                cross_age_bins + 1
            )
            age_edges = (
                width2_edges - sigma_injection2
            ) / (2 * kt25_kappa)
            for bin_index, (lower, upper) in enumerate(zip(
                age_edges[:-1], age_edges[1:]
            )):
                selected = (
                    valid_return
                    & ~exact_selected
                    & (time_since_return >= lower)
                    & (time_since_return < upper)
                )
                if not np.any(selected):
                    continue
                width2_midpoint = np.sqrt(
                    width2_edges[bin_index]
                    * width2_edges[bin_index + 1]
                )
                sigma = np.sqrt(width2_midpoint)
                fields[element_index] += _deposit_binned_gaussian(
                    return_x[selected],
                    return_y[selected],
                    event_yield[
                        selected, element_index, channel_index
                    ],
                    sigma
                )

    residual = fields - cross_axisymmetric_mean
    residual[:, ~cross_disk_mask] = np.nan
    values = residual[:, cross_analysis_mask]
    covariance = values @ values.T / values.shape[1]
    # Retain only O and N maps between realizations. The full element set has
    # already contributed to the covariance matrices above.
    return {
        'fields': fields[cross_saved_index].astype(np.float32),
        'residual': residual[cross_saved_index].astype(np.float32),
        'covariance': covariance,
        'n_event': n_event,
    }


cross_validation_cache_version = 4
cross_validation_seed = 20260625
cross_validation_cache_path = (
    NOTEBOOK_ROOT / 'data'
    / f'{validation_output_prefix}validation_cross_cache.npz'
)
cross_validation_checkpoint_path = (
    NOTEBOOK_ROOT / 'data'
    / f'{validation_output_prefix}validation_cross_checkpoint.npz'
)


def _cross_validation_signature():
    """Fingerprint every input that changes the cached realizations."""
    import hashlib

    digest = hashlib.sha256()
    signature_parts = [
        np.asarray(cross_validation_elements, dtype='S'),
        np.asarray([
            cross_validation_n_realizations,
            cross_age_bins,
            cross_exact_sigma_threshold,
            cross_validation_seed,
            cross_dx,
            cross_source_padding,
            R_in,
            R_out,
            h_R,
            m_arms,
            pitch_angle,
            Theta0,
            eta,
            float(enable_source_clustering),
            cluster_fraction,
            cluster_radius,
            mean_children_per_cluster,
            kt25_t_corr,
            kt25_Gamma,
            kt25_kappa,
            kt25_sigma_v,
            kt25_event_mass_mean,
            float(slug_moments['cmf_min']),
            float(slug_moments['cmf_max']),
        ], dtype=np.float64),
        harmonic_n,
        harmonic_g,
        harmonic_alpha,
        kt25_sigma_inj,
        cross_mean_yield,
        cross_delay,
        cross_K,
    ]
    for value in signature_parts:
        array = np.ascontiguousarray(value)
        digest.update(str(array.dtype).encode('ascii'))
        digest.update(np.asarray(array.shape, dtype=np.int64).tobytes())
        digest.update(array.tobytes())
    return digest.hexdigest()


cross_validation_signature = _cross_validation_signature()


def _load_cross_validation_cache(path, require_complete):
    """Load a compatible complete cache or an interrupted checkpoint."""
    if not path.exists():
        return None
    try:
        with np.load(path, allow_pickle=False) as cached:
            version = int(cached['cache_version'])
            signature = str(cached['signature'])
            covariance_stack = cached['covariance_stack'].copy()
            if version != cross_validation_cache_version:
                return None
            if signature != cross_validation_signature:
                return None
            if covariance_stack.ndim != 3:
                return None
            if require_complete and (
                len(covariance_stack)
                != cross_validation_n_realizations
            ):
                return None
            if len(covariance_stack) > cross_validation_n_realizations:
                return None

            display_realization = {
                'fields': cached['display_fields'].copy(),
                'residual': cached['display_residual'].copy(),
                'covariance': cached['display_covariance'].copy(),
                'n_event': int(cached['display_n_event']),
            }
            rng_state = (
                str(cached['rng_state'])
                if 'rng_state' in cached.files
                else None
            )
    except (KeyError, OSError, ValueError):
        return None

    return covariance_stack, display_realization, rng_state


def _save_cross_validation_cache(
    path, covariance_stack, display_realization, rng
):
    """Write a restartable checkpoint atomically."""
    import json

    path.parent.mkdir(parents=True, exist_ok=True)
    temporary_path = path.with_suffix(path.suffix + '.tmp')
    with temporary_path.open('wb') as stream:
        np.savez_compressed(
            stream,
            cache_version=np.array(
                cross_validation_cache_version, dtype=np.int64
            ),
            signature=np.array(cross_validation_signature),
            covariance_stack=np.asarray(
                covariance_stack, dtype=np.float64
            ),
            display_fields=np.asarray(
                display_realization['fields'], dtype=np.float32
            ),
            display_residual=np.asarray(
                display_realization['residual'], dtype=np.float32
            ),
            display_covariance=np.asarray(
                display_realization['covariance'], dtype=np.float64
            ),
            display_n_event=np.array(
                display_realization['n_event'], dtype=np.int64
            ),
            rng_state=np.array(json.dumps(rng.bit_generator.state)),
        )
    temporary_path.replace(path)


cached_validation = _load_cross_validation_cache(
    cross_validation_cache_path, require_complete=True
)
if cached_validation is not None:
    (
        cross_covariance_stack,
        cross_display_realization,
        _,
    ) = cached_validation
    cross_validation_realizations = [cross_display_realization]
    print(
        f'Loaded {len(cross_covariance_stack)} KTZ cross realizations '
        f'from {cross_validation_cache_path.name}'
    )
else:
    import json

    cross_validation_rng = default_rng(cross_validation_seed)
    checkpoint = _load_cross_validation_cache(
        cross_validation_checkpoint_path, require_complete=False
    )
    if checkpoint is None:
        cross_covariances = []
        cross_display_realization = None
    else:
        (
            checkpoint_stack,
            cross_display_realization,
            checkpoint_rng_state,
        ) = checkpoint
        cross_covariances = list(checkpoint_stack)
        if checkpoint_rng_state is not None:
            cross_validation_rng.bit_generator.state = json.loads(
                checkpoint_rng_state
            )
        print(
            f'Resuming after {len(cross_covariances)}/'
            f'{cross_validation_n_realizations} KTZ cross realizations'
        )

    try:
        for realization_index in range(
            len(cross_covariances),
            cross_validation_n_realizations
        ):
            realization = generate_cross_ktz_realization(
                cross_validation_rng
            )
            cross_covariances.append(realization['covariance'])
            if cross_display_realization is None:
                cross_display_realization = realization
            _save_cross_validation_cache(
                cross_validation_checkpoint_path,
                cross_covariances,
                cross_display_realization,
                cross_validation_rng,
            )
            print(
                f'KTZ cross realization {realization_index + 1}/'
                f'{cross_validation_n_realizations}: '
                f'{realization["n_event"]:,} sampled parent events'
            )
    except KeyboardInterrupt:
        print(
            '\nInterrupted after the last completed realization. '
            'Re-run this cell to resume from the saved checkpoint.'
        )
        raise

    cross_covariance_stack = np.stack(cross_covariances)
    cross_validation_realizations = [cross_display_realization]
    _save_cross_validation_cache(
        cross_validation_cache_path,
        cross_covariance_stack,
        cross_display_realization,
        cross_validation_rng,
    )
    cross_validation_checkpoint_path.unlink(missing_ok=True)

cross_simulation_covariance = np.mean(
    cross_covariance_stack, axis=0
)
# KTZ predicts a Pearson coefficient constructed from ensemble covariance:
# E[C_XY] / sqrt(E[C_XX] E[C_YY]), not the mean of separately normalized
# single-realization Pearson coefficients.
cross_simulation_mean = normalize_cross_covariance(
    cross_simulation_covariance
)

cross_bootstrap_count = 1000
cross_bootstrap_rng = default_rng(20260626)
cross_bootstrap_correlation = np.empty(
    (
        cross_bootstrap_count,
        len(cross_validation_elements),
        len(cross_validation_elements)
    )
)
for bootstrap_index in range(cross_bootstrap_count):
    selection = cross_bootstrap_rng.integers(
        0, cross_validation_n_realizations,
        cross_validation_n_realizations
    )
    covariance = np.mean(
        cross_covariance_stack[selection], axis=0
    )
    cross_bootstrap_correlation[bootstrap_index] = (
        normalize_cross_covariance(covariance)
    )
# Bootstrap standard error of the ensemble-mean coefficient; this is not
# the realization-to-realization scatter of separately normalized maps.
cross_simulation_mean_error = np.std(
    cross_bootstrap_correlation, axis=0, ddof=1
)
cross_square_pixel_cp0 = kt25_square_pixel_covariance_matrix(
    cross_dx, kt25_second_yield_normalized
)
cross_pixel_components = compute_ktz_window_components(
    kt25_sigma_inj, pixel_size=cross_dx,
    cp0_override=cross_square_pixel_cp0
)
cross_pixel_covariance = sum(cross_pixel_components.values())
Xi_KTZ_pixel = normalize_cross_covariance(
    cross_pixel_covariance
)
cross_theory_subset = Xi_KTZ_pixel[
    np.ix_(cross_validation_index, cross_validation_index)
]
print(
    f'Cross simulation: {cross_expected_parent_events:,.0f} '
    f'events per realization, '
    f'{1000 * cross_dx:.0f} pc pixels'
)
```
